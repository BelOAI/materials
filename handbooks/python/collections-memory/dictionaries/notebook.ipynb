## Значение по ключу

Словарь `dict` связывает уникальный ключ со значением. Для строки данных ключом может быть название поля, для частот — слово. Словарь сохраняет порядок добавления ключей, но обращение идёт по ключу, а не по позиции. Оператор `in` проверяет наличие ключа.

```python
row = {"id": "g01", "span_cm": 12}
assert row["span_cm"] == 12
row["mass_g"] = 8
assert "mass_g" in row
```

Повторное присваивание по одному ключу заменяет прежнее значение. Обращение к отсутствующему ключу через квадратные скобки вызывает `KeyError`. `get(key, default)` возвращает значение ключа, а если ключа нет — значение `default`. Новый ключ при этом не добавляется.

## Разобранный пример: частоты слов

```python
counts = {}
for word in "кот кит кот".split():
    counts[word] = counts.get(word, 0) + 1
assert counts == {"кот": 2, "кит": 1}
```

При первом слове «кот» метод `get` возвращает 0, затем записывается 1. Во второй раз возвращается 1, записывается 2. Для пустого текста цикл не выполнится, результатом будет пустой словарь.

## Обход и включения

Обычный цикл по словарю перебирает ключи. Методы `values()` и `items()` дают значения и пары «ключ, значение». Во включении перед двоеточием записывается ключ, после — значение.

```python
squares = {number: number ** 2 for number in [2, 3]}
assert squares == {2: 4, 3: 9}
total = 0
for word, count in counts.items():
    total += count
assert total == 3
```

Ключами могут быть числа, строки и кортежи подходящих неизменяемых элементов. Списки ключами быть не могут. Для собственной таблицы проверяйте уникальность ключа до записи: два объекта с одним id иначе незаметно заменят друг друга.

## Как выбирать значение по отсутствующему ключу

`get` не меняет словарь: `counts.get("новое", 0)` только возвращает 0. Чтобы сохранить счётчик, нужно присваивание `counts["новое"] = ...`. Включение `{number: number ** 2 for number in [2, 3]}` делает две записи: сначала ключ 2 со значением 4, затем ключ 3 со значением 9. Как и у спискового включения, результат создаётся заново.
