## Аргументы и результат

Параметры — имена в определении функции; аргументы — значения при вызове. Аргументы можно передать по позиции или по имени. Значение по умолчанию используется, если аргумент пропущен.

`sum` суммирует элементы, тройные кавычки задают строку документации. `return` немедленно завершает функцию; без него результат равен `None`. `print` показывает текст, но не заменяет возвращаемое значение.

```python
def count_above(values, threshold=20):
    """Число элементов не меньше порога."""
    return sum([1 for value in values if value >= threshold])

print(count_above([12, 22, 24]))                 # 2
print(count_above([12, 22, 24], threshold=23))   # 1
```

Имена, созданные внутри функции, локальны: снаружи к ним нельзя обратиться по этим именам. Внешнее имя можно прочитать, но лучше передать нужные данные аргументами — тогда поведение проще проверить.

```python
threshold = 20
def shifted(value):
    threshold = 5
    return value + threshold

print(shifted(2), threshold)  # 7 20
```

## Изменяемые аргументы

Функция получает ссылку на объект. Изменение переданного списка через `append` видно вызывающему коду. Для функции, которая должна сохранять вход, создайте новый список.

Значения по умолчанию вычисляются один раз при определении функции. Поэтому `def add(value, items=[])` опасно: вызовы будут делить один список. Используйте `None`:

```python
def add(value, items=None):
    if items is None:
        items = []
    items.append(value)
    return items

assert add(1) == [1]
assert add(2) == [2]
```

## Ошибки и трассировка

`SyntaxError` означает, что запись программы не удалось разобрать. Во время выполнения возможны `NameError` (неизвестное имя), `TypeError` (неподходящий тип), `ValueError` (недопустимое значение), `IndexError` и `KeyError`.

Трассировка — список вызовов функций, приведших к ошибке, с указанием файлов и строк. В ней сначала прочитайте последнюю строку: тип и сообщение. Затем найдите последнюю строку своего кода и проследите вызовы выше. Например, `int("крыло")` приводит к `ValueError`: строку нельзя прочитать как целое число.

`try` выполняет код, `except ValueError` обрабатывает только исключение типа `ValueError`. Не перехватывайте все ошибки, чтобы просто продолжить работу: потерянные строки данных изменят результат незаметно. Там, где неверный вход нельзя исправить, явно сообщите о нём через `raise`.

```python
try:
    number = int("крыло")
except ValueError:
    print("Не удалось прочитать целое число")
```

## Разобранный пример: среднее с проверкой

`assert` проверяет утверждение и вызывает `AssertionError`, если оно ложно. `pass` означает «ничего не делать», а `else` после `try` выполняется, если исключения не было. Для проверки ошибок важно убедиться, что исключение действительно возникло. `assert` подходит для учебных тестов, но проверку пользовательских данных делайте через `if` и `raise`: при оптимизированном запуске Python утверждения могут отключаться.

```python
def average(values):
    if len(values) == 0:
        raise ValueError("Среднее пустого набора не определено")
    return sum(values) / len(values)

assert average([10, 14, 18]) == 14
assert average([7]) == 7
try:
    average([])
except ValueError:
    pass
else:
    raise AssertionError("Пустой вход должен вызвать ValueError")
```

Встроенные `min(values)` и `max(values)` находят наименьшее и наибольшее значения непустого набора. У пустого набора минимума и максимума нет. Проверку пустоты списка можно записать `if len(values) == 0` или короче `if not values`: пустой список считается ложным.

Разделите чтение данных, проверку и вычисление на разные функции. Тогда ошибку можно искать на маленьком примере, не запуская весь эксперимент.
