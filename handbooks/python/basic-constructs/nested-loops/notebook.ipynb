## Как выполняются два цикла

Внутренний цикл начинается заново для каждого значения внешнего. Для прямоугольника из двух строк и трёх столбцов получаем шесть ячеек. Пара `(row, column)` хранит два значения вместе; такая неизменяемая последовательность называется кортежем. Метод — действие, вызываемое через точку после объекта. `cells.append(...)` добавляет элемент в конец списка `cells`.

При `row=0` внутренний цикл перебирает `column=0, 1, 2`. Затем `row` становится 1, а внутренний цикл снова начинает с 0. Поэтому получаются 2 · 3 пары. Вложенное тело имеет два уровня отступа.

```python
cells = []
for row in range(2):
    for column in range(3):
        cells.append((row, column))
assert len(cells) == 6
print(cells)  # [(0, 0), (0, 1), (0, 2), (1, 0), (1, 1), (1, 2)]
```

`break` завершает только ближайший цикл. Если нужно прекратить весь поиск, можно сохранить логическое значение `found` и проверить его во внешнем цикле. Не изменяйте список, по которому идёт перебор: количество и положение элементов могут поменяться.

## Разобранный пример: пары подходящих размахов

Нужны пары разных объектов с суммой размахов 34. Метод `pairs.append(...)` добавит пару индексов в список результатов. Для индексов `i < j` каждая пара рассматривается ровно один раз.

```python
spans = [12, 22, 14, 20]
pairs = []
for i in range(len(spans)):
    for j in range(i + 1, len(spans)):
        if spans[i] + spans[j] == 34:
            pairs.append((i, j))
print(pairs)  # [(0, 1), (2, 3)]
```

Подробнее пары значений разберём в коллекциях. Например, при `i = 0` внутренний цикл проверяет `j = 1, 2, 3`; при `i = 1` — только `j = 2, 3`. Перебор всех пар может быть дорогим — оценивать его стоимость научимся в главе об алгоритмах.

## Разобранный пример: найти хотя бы одну пару

Если нужен только факт существования пары, не обязательно сохранять все ответы. При нахождении пары завершим оба цикла.

```python
values = [4, 7, 2]
found = False
for i in range(len(values)):
    for j in range(i + 1, len(values)):
        if values[i] + values[j] == 9:
            found = True
            break
    if found:
        break
assert found
```

Проверьте список из одного элемента: пары разных объектов быть не может. Если два объекта имеют одинаковое значение, это всё равно разные объекты: `[5, 5]` содержит пару с суммой 10, а `[5]` — нет. Поэтому перебираем индексы, а не только разные значения.
