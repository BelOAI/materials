## От вопроса к условию

Условие отвечает на вопрос с двумя ответами: «да» или «нет». Например, `mass < 15` спрашивает, меньше ли масса 15. Результат `True` означает «да», `False` — «нет». В отличие от присваивания `mass = 12`, сравнение `mass == 12` не меняет массу.

После условия `if` ставят двоеточие, а команды выбранной ветви сдвигают вправо на четыре пробела. Одинаковые отступы объединяют команды в блок. Строка без этого отступа после ветвей выполняется независимо от выбора.

## Логические выражения

Сравнения `==`, `!=`, `<`, `<=`, `>`, `>=` возвращают `True` или `False`. `and` требует выполнения обоих условий, `or` — хотя бы одного, `not` меняет истину на ложь и наоборот. Скобки помогают явно задать порядок.

```python
mass = 12
print(10 <= mass < 15)                     # True
print(mass >= 10 and mass < 15)            # True
print(not (mass < 10 or mass >= 15))       # True
```

`and` не вычисляет правую часть, если левая уже ложна. Поэтому проверка `count != 0 and total / count > 10` безопасна и при нулевом `count`: деление тогда не выполняется. Обратный порядок такой защиты не даёт.

## Выбор одной ветви

В конструкции `if` / `elif` / `else` выполняется только первая подходящая ветвь. `elif` означает «иначе, если». Независимые `if` могут выполнить несколько действий.

```python
score = 70
if score >= 90:
    level = "отлично"
elif score >= 60:
    level = "зачёт"
else:
    level = "нужна практика"
print(level)  # зачёт
```

Ставьте более узкое условие раньше широкого: при проверке `score >= 60` первой оценка 95 тоже попадёт в «зачёт».

## Разобранный пример: безопасное деление

В программе есть два числа — сумма и количество измерений. Деление допустимо только при ненулевом количестве. Проверку нужно выполнить до деления. Двоеточие начинает блок; вложенные строки имеют отступ четыре пробела.

```python
total = 42
count = 3
if count > 0:
    print(total / count)  # 14.0
else:
    print("Нет измерений")
```

Теперь замените `count` на 0: программа должна вывести сообщение. Если поставить вычисление `total / count` перед `if`, условие уже не сможет защитить от ошибки. Для нескольких проверок `and` требует истинности обеих, `or` — хотя бы одной. Скобки помогают показать замысел явно.

## Границы и порядок проверок

Условия полезно проверять на границе и рядом с ней. Для порога 60 возьмите 59, 60, 61. Для промежутка от 10 до 15 включительно — 9, 10, 15, 16. При нескольких независимых `if` возможны несколько сообщений, а в цепочке `if`–`elif`–`else` выбирается одна ветвь.

## Что считается истиной

В условии можно использовать не только сравнение. Число 0 и пустая строка считаются ложными, ненулевое число и непустая строка — истинными. Однако при проверке равенства пишите каждое сравнение полностью: `x == 1 or x == 2`. Запись `or 2` оставляет справа самостоятельное ненулевое число, а не второе сравнение.
