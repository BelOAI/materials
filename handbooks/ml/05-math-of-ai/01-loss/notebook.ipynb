## Добавим один крупный промах

Меняйте последнее предсказание A. Отметьте, при каком значении порядок моделей по MAE и MSE различается.

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error
truth = np.array([20, 30, 40, 50, 60])
a = np.array([18, 29, 41, 50, 60])
b = np.array([17, 27, 37, 47, 57])
for last in [60, 68, 76]:
    changed = a.copy(); changed[-1] = last
    print(last, 'A:', round(mean_absolute_error(truth, changed),2), round(mean_squared_error(truth, changed),2), 'B:', mean_absolute_error(truth,b), mean_squared_error(truth,b))
```

## Уверенная ошибка

Два классификатора выбирают одинаковые классы. Сравните кросс-энтропию, когда один из них увереннее ошибается.

```python
from sklearn.metrics import log_loss
labels = np.array([1,0,1,0])
p_a = np.array([0.8,0.2,0.4,0.1])
p_b = np.array([0.8,0.2,0.01,0.1])
for name, p in [('A',p_a),('B',p_b)]:
    print(name, 'доля верных', np.mean((p>=0.5)==labels), 'кросс-энтропия', round(log_loss(labels,p),3))
```
