# Практика § 5.2: градиент и размер шага

Сначала посчитайте знак изменения веса. Здесь рассматривается только учебный нейрон; результаты не доказывают свойства глубоких сетей.

## Численный градиент

Поменяйте размер приращения и сравните приближение с точным значением −2.

```python
def loss_at(w, b=0.0):
    pred = max(0.0, 2*w+b)
    return (pred-2)**2/2
for eps in [0.1, 0.01, 0.001]:
    numerical = (loss_at(0.5+eps)-loss_at(0.5-eps))/(2*eps)
    print(eps, round(numerical,5))
```

## Autograd

Проверьте отдельно градиент веса и смещения.

```python
import torch
w = torch.tensor(0.5, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
loss = (torch.relu(2*w+b)-2).square()/2
loss.backward()
print(float(w.grad), float(b.grad))
```

## Несколько размеров шага

Постройте таблицу одной и той же начальной точки. На какой шаг улучшение уже не распространяется?

```python
for eta in [0.01, 0.1, 0.4, 1.0]:
    new_w = 0.5 - eta*(-2)
    new_b = 0.0 - eta*(-1)
    print(eta, round(new_w,3), round(new_b,3), round(loss_at(new_w,new_b),4))
```
