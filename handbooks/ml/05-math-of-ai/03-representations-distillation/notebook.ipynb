## Косинус и масштаб

Умножьте последний вектор на 10. Что происходит со скалярным произведением и косинусом?

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
query = np.array([[1.,0.]])
images = np.array([[1.,0.],[0.,1.],[.8,.6]])
for scale in [1,10]:
    changed = images.copy(); changed[-1] *= scale
    print(scale, 'скалярные', query @ changed.T, 'косинусы', cosine_similarity(query,changed).round(3))
```

## Мягкие цели

Сравните распределения и порядок. Поменяйте местами две последние оценки учителя.

```python
def softmax(z,t):
    z=np.asarray(z,dtype=float)/t
    a=np.exp(z-z.max())
    return a/a.sum()
logits=np.array([3.,2.,0.])
for t in [1.,2.,5.]:
    print(t, softmax(logits,t).round(3), 'первый', softmax(logits,t).argmax())
```

## Игрушечный ученик

Один набор логитов ученика не является обученной моделью для новых запросов. Сравните, насколько его доли близки к учителю при изменении параметра.

```python
target=softmax(logits,2.)
for student_logits in [[0.,0.,0.],[1.,.5,0.],[2.,1.,0.]]:
    predicted=softmax(student_logits,2.)
    cross_entropy=-np.sum(target*np.log(predicted))
    print(student_logits, round(cross_entropy,3))
```
