## Одни значения, разные контексты

Измените оценки для второго контекста. Что случится с итоговым числом, если больший вес получит значение 2?

```python
import numpy as np
values = np.array([1.0, 0.0, 2.0])
def attend(scores):
    scores = np.asarray(scores, dtype=float)
    weights = np.exp(scores - scores.max())
    weights /= weights.sum()
    return weights, weights @ values
for scores in ([2,1,0], [0,1,2]):
    weights, result = attend(scores)
    print(np.round(weights,3), round(float(result),3))
```
