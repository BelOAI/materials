# Практика § 4.4: причинная маска

Код показывает доступ к уже написанным токенам. Он не обучает большую языковую модель.

## Постройте маску

Предскажите число открытых клеток для длины 5. Затем проверьте кодом.

```python
import numpy as np
n = 5
allowed = np.tril(np.ones((n,n), dtype=bool))
print(allowed.astype(int))
print("Открытых клеток:", allowed.sum())
```

## Запретите будущий токен

В четвёртой позиции оценки последних двух будущих токенов велики. Сравните неправильную нормировку по всей строке и правильную по доступному префиксу.

```python
scores = np.array([0.,1.,2.,3.,8.])
position = 2
def softmax(x):
    x = np.asarray(x)
    w = np.exp(x-x.max())
    return w/w.sum()
print("Неправильно, вся строка:", np.round(softmax(scores),3))
print("Правильно, префикс:", np.round(softmax(scores[:position+1]),3))
```

## Обучающие пары

Вставьте собственную короткую фразу. Проверьте, что число целей на одну меньше числа токенов, если первый токен служит началом.

```python
tokens = ["<начало>", "В", "кружке", "печатают", "фигурку", "<конец>"]
for i in range(1,len(tokens)):
    print(tokens[:i], "→", tokens[i])
```
