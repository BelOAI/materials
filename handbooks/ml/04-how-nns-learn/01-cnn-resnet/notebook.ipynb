Сначала предскажите результат. Числовые признаки ниже придуманы; они показывают обучение классификатора при замороженном извлекателе, но не являются выходом ResNet.

## Остаточный блок

Измените поправку, затем объясните, какая координата обнулилась после ReLU. Найдите случай, когда обе координаты сохраняются.

```python
import numpy as np
x = np.array([1.0, -2.0])
correction = np.array([-0.4, 0.5])
print("До активации:", x + correction)
print("После ReLU:", np.maximum(x + correction, 0))
```

## Учим только классификатор

Предположите, как новый двумерный признак влияет на класс. Векторы фиксированы: мы обновим только веса логистической регрессии. Для настоящих признаков ResNet потребуется извлечь их из размеченных снимков.

```python
from sklearn.linear_model import LogisticRegression
features = np.array([[2, 0], [1.8, 0.2], [2.2, 0.1],
                     [0, 2], [0.2, 1.8], [0.1, 2.2]], dtype=float)
labels = np.array([0, 0, 0, 1, 1, 1])
head = LogisticRegression().fit(features, labels)
print("Вес новой головки:", head.coef_.round(2))
print("Прогноз:", head.predict([[1.7, 0.3], [0.3, 1.7]]))
```
