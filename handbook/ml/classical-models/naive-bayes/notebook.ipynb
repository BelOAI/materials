## Упрощение совместной модели

Чтобы учесть много признаков, наивный Байес предполагает их независимость **при заданном классе**: P(x₁,…,xₚ|y)=произведение P(xⱼ|y). Это не утверждение, что признаки независимы вообще. Сначала выбирается класс, затем в его рамках используется приближённое разложение.

Для бинарных признаков, например наличия слова, учитываются и присутствие, и отсутствие. Для счётчиков слов часто применяют мультиномиальную модель; для числовых признаков — гауссовскую. Эти варианты делают разные предположения о данных.

## Разобранный пример: два слова

Классы равновероятны. При классе А слова «спорт» и «матч» присутствуют с шансами 0.8 и 0.6, при Б — 0.2 и 0.1. Для сообщения с обоими словами веса 0.5·0.8·0.6=0.24 и 0.5·0.2·0.1=0.01. После нормировки вероятность А равна 0.24/0.25=0.96.

Но «спорт» и «матч» могут быть связаны даже внутри А. Если два столбца фактически копируют один сигнал, перемножение учитывает его дважды и делает прогноз слишком уверенным. Хорошие метки не гарантируют калиброванные вероятности.

## Нули и сглаживание

Если слово ни разу не встретилось в одном классе, несглаженная оценка вероятности ноль обнулит всё произведение. Добавление alpha к счётчикам, например alpha=1, предотвращает абсолютные запреты на основании маленькой выборки. В sklearn BernoulliNB подходит для бинарных признаков; параметр alpha задаёт сглаживание, predict_proba — итоговые вероятности.

Например, в классе четыре документа, и слово встретилось в одном. Без сглаживания оценка вероятности присутствия равна 1/4. При alpha=1 добавим единицу к обоим бинарным исходам: присутствию и отсутствию. Получим $(1+1)/(4+2)=1/3$, для отсутствия — $(3+1)/(4+2)=2/3$. Сумма остаётся 1. Если слово не встретилось вовсе, его оценка будет $1/(4+2)=1/6$, а не ноль.

Произведения многих малых чисел могут численно обнулиться. Реализации обычно суммируют логарифмы и нормируют устойчиво. Это та же модель, записанная в удобной численной форме.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

Для библиотечного опыта используем другую маленькую таблицу: по два документа каждого класса, два бинарных признака. Сглаживание alpha=1 даёт для первого признака вероятности 3/4 в классе 1 и 1/4 в классе 0; для второго — 1/2 в обоих. При равных долях классов объект [1,1] получает P(класс 1)=0.75. Это вычисление относится к этой таблице; в ручном примере выше вероятности были заданы отдельно.

```python
import numpy as np
from sklearn.naive_bayes import BernoulliNB
X = np.array([[1, 1], [1, 0], [0, 0], [0, 1]])
y = np.array([1, 1, 0, 0])
model = BernoulliNB(alpha=1.).fit(X, y)
p = model.predict_proba([[1, 1]])
assert p.shape == (1, 2) and np.allclose(p.sum(axis=1), 1)
assert np.all(p > 0)
assert model.classes_.tolist() == [0, 1]
assert np.allclose(p, [[0.25, 0.75]])
print(model.classes_, p)
```
