## Чему научимся

Объяснять, как bootstrap и случайный выбор признаков создают разнообразие деревьев.

## Несколько нестабильных решений

Глубокое дерево может сильно меняться от небольшой замены объектов в train. **Bagging** обучает несколько моделей на разных повторных выборках и объединяет ответы. Для регрессии часто усредняют числа, для классификации — голоса или вероятности. Уменьшение разброса сильнее, если ошибки моделей не полностью одинаковы.

Bootstrap-выборка имеет тот же размер n, что train, но строится с возвращением: некоторые объекты повторяются, некоторые не попадают. Это не новый независимый набор измерений. Проверка исходных данных по-прежнему нужна.

## Разобранный пример: усреднение

Три регрессора дали одному объекту 8,10,12. Средний прогноз 10. Если один из них ошибся вверх, а другой вниз, ошибки частично компенсируются. Но если все обучены на утечке или все систематически завышают ответ на 5, усреднение проблему не исправит.

**Случайный лес** добавляет к bootstrap случайное подмножество признаков при поиске разбиения узла. Деревья меньше повторяют одни и те же сильные вопросы, разнообразие растёт. Параметры n_estimators и max_features задают число деревьев и доступные признаки при разбиении. Увеличение числа деревьев обычно стабилизирует среднее, но увеличивает время и память.

## Проверка ансамбля

Объекты, не попавшие в bootstrap конкретного дерева, называют out-of-bag. Они позволяют получать внутреннюю оценку при подходящих предположениях, но не заменяют групповую или временную проверку. Другая фотография того же человека всё ещё может быть в обучении.

В ноутбуке используем небольшой лес и явный seed. Повторение при одинаковых данных и настройках проверяет воспроизводимость, а не доказывает качество на настоящей задаче.

`estimators_` содержит обученные деревья. `n_jobs=1` ограничивает параллельность одного запуска; это не настройка математического критерия.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
X = [[float(i)] for i in range(12)]
y = [0] * 6 + [1] * 6
first = RandomForestClassifier(n_estimators=12, max_depth=3, random_state=17, n_jobs=1).fit(X, y)
second = RandomForestClassifier(n_estimators=12, max_depth=3, random_state=17, n_jobs=1).fit(X, y)
assert np.array_equal(first.predict(X), second.predict(X))
assert len(first.estimators_) == 12
print(first.predict([[1.5], [9.5]]))
```
