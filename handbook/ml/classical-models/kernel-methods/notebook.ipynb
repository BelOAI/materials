## Нелинейность через новое описание

Иногда классы не разделяются прямой в исходных координатах. Можно добавить признаки: для x взять φ(x)=(x,x²). Тогда линейная граница в новом описании станет нелинейной по исходному x. Но при большом числе новых признаков явное хранение неудобно.

**Ядро** K(x,z)=φ(x)·φ(z) вычисляет скалярное произведение новых описаний, не обязательно создавая их явно. Не всякая произвольная функция сходства является допустимым ядром SVM; в этом уроке используем стандартные проверенные варианты.

Одной симметрии недостаточно уже потому, что $K(x,x)=\varphi(x)\cdot\varphi(x)=\|\varphi(x)\|^2\geq0$. Симметричная таблица с −1 на диагонали не может быть таблицей таких произведений. Даже неотрицательная диагональ сама по себе ещё не гарантирует допустимость; поэтому здесь не конструируем ядро из произвольных оценок сходства.

## Разобранный пример: квадрат суммы

Пусть φ(x)=(1,√2·x,x²) для одного числа. Тогда φ(x)·φ(z)=1+2xz+x²z²=(1+xz)². Для x=2,z=3 получаем 49 обоими способами. Это полиномиальное ядро второй степени.

RBF-ядро K(x,z)=exp(−γ||x−z||²) при γ>0 даёт 1 для одинаковых точек и меньшие значения для далёких. Большой γ делает влияние объекта более локальным; очень локальная модель может запоминать шум. Параметр C управляет регуляризацией, поэтому γ и C подбирают вместе по внутренней проверке.

## Цена попарных сравнений

Для n объектов плотная таблица ядер имеет n² чисел. При n=10 000 и восьми байтах на число это 800 миллионов байт только на элементы, без копий и рабочих структур. Фраза «не создаём признаки явно» не означает бесплатное обучение.

Масштабирование исходных координат влияет на расстояния RBF. Параметры масштабирования вычисляют по train; в конвейере этот шаг ставят перед расчётом ядра. В ноутбуке сначала вручную проверяем равенство двух способов расчёта, затем создаём небольшую таблицу ядра; это не оценка скорости на большом наборе.

`rbf_kernel` строит попарную таблицу указанного ядра. `np.diag(K)` выбирает её диагональ, чтобы проверить сравнение каждой точки с собой.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.metrics.pairwise import rbf_kernel
x, z = 2., 3.
phi_x = np.array([1., np.sqrt(2) * x, x*x])
phi_z = np.array([1., np.sqrt(2) * z, z*z])
assert np.isclose(phi_x @ phi_z, (1 + x*z)**2)
K = rbf_kernel([[0.], [1.], [2.]], gamma=0.5)
assert K.shape == (3, 3)
assert np.allclose(K, K.T) and np.allclose(np.diag(K), 1)
print(K)
```
