## Чему научимся

Получать вероятности нескольких классов из счётов и учитывать инвариантность softmax.

## Несколько возможных ответов

Для трёх классов модель выдаёт три счёта z₁,z₂,z₃. **Softmax** превращает их в вероятности: $p_k=e^{z_k}/\sum_j e^{z_j}$. Все значения положительны и суммируются в 1. Больший счёт даёт большую вероятность, но размер зависит от остальных счётов тоже.

Многоклассовая кросс-энтропия одного объекта равна −ln p правильного класса. Не нужно отдельно штрафовать каждый неверный класс как независимый ответ: в этом уроке классы взаимоисключающие. Задача с несколькими одновременно допустимыми метками требует другой постановки.

## Разобранный пример: пропорции 1:2:3

Возьмём счёты [0,ln2,ln3]. Экспоненты равны [1,2,3], сумма 6, вероятности [1/6,2/6,3/6]. Победит третий класс, но его вероятность всего 0.5. Метка-победитель не сообщает всю неопределённость.

Если ко всем счётам прибавить c, экспоненты умножатся на e^c, которое сократится. Поэтому softmax(z+c)=softmax(z). Для устойчивого вычисления обычно вычитают максимальный счёт: положительные огромные экспоненты исчезают, а результат сохраняется. Это не меняет порядок классов.

## Температура

Разделение счётов на положительное число T называют температурой. При большем T распределение становится ровнее, при меньшем — острее; порядок победителей сохраняется при отсутствии равенств. Температура не создаёт новых знаний и не гарантирует калибровку, если просто выбрана на глаз.

В коде np.exp вычисляет экспоненту поэлементно, sum(axis=1,keepdims=True) сохраняет ось столбцов для деления каждой строки на свою сумму. До потери проверяйте, какой столбец соответствует правильной метке.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
scores = np.array([[0., np.log(2), np.log(3)], [1000., 1000., 1000.]])
shifted = scores - scores.max(axis=1, keepdims=True)
weights = np.exp(shifted)
p = weights / weights.sum(axis=1, keepdims=True)
assert np.allclose(p.sum(axis=1), 1)
assert np.allclose(p[0], [1/6, 2/6, 3/6])
assert np.allclose(p[1], [1/3, 1/3, 1/3])
print(p)
```
