## Две причины плохого прогноза

Прямая линия плохо описывает выраженно изогнутую зависимость даже на train: это возможное **недообучение**. Очень гибкая кривая может проходить через все шумные точки train, но сильно отклоняться между ними: это **переобучение**. Увеличение сложности обычно уменьшает обучающий штраф, но не обязано уменьшать проверочный.

Смещение описывает систематическую ошибку среднего прогноза по возможным обучающим выборкам. Разброс модели — изменение её прогноза от выборки к выборке. Это не то же самое, что дисперсия целевого признака. Шум ответа может оставаться даже при правильной модели.

## Разобранный пример: две таблицы качества

У модели А MSE train=10, validation=11. У Б train=0.1, validation=20. А может быть слишком простой, Б — подстраиваться под конкретные точки. Однако одних двух чисел мало: проверьте отсутствие сдвига выборок, одинаковые единицы и достаточность данных. Утечка иногда даёт подозрительно хорошую оценку, а неверная проверка — ложный диагноз.

**Кривая обучения** сравнивает качество при разном размере train. Если разрыв сокращается при добавлении данных, это аргумент в пользу снижения разброса. **Кривая сложности** сравнивает настройки при одном протоколе. Их нельзя строить по многократному просмотру итогового теста.

В численном примере несколько разных наборов наблюдений порождают разные прогнозы в одной точке. Мы вычисляем средний прогноз и разброс его отклонений. Это наглядная демонстрация понятий, а не доказательство точного разложения ошибки любой метрики.

Пусть три обучения дали в одной и той же точке прогнозы 2,4,6. Их центр $(2+4+6)/3=4$, средний квадрат отклонения от центра $(4+0+4)/3=8/3$. Если истинное условное среднее в этой точке равно 5, смещение среднего прогноза равно $4-5=-1$. Разброс и смещение отвечают на разные вопросы: насколько прогнозы меняются и куда в среднем промахиваются.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
predictions = np.array([2., 4., 6.])
center = predictions.mean()
spread = ((predictions - center) ** 2).mean()
assert center == 4
assert np.isclose(spread, 8 / 3)
# Условные оценки с одного и того же протокола; не результаты обученной здесь модели.
scores = {"A": {"train_mse": 10., "valid_mse": 11.}, "B": {"train_mse": 0.1, "valid_mse": 20.}}
assert scores["B"]["valid_mse"] - scores["B"]["train_mse"] > 19
print(center, spread)
```
