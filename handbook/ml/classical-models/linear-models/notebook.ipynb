## Модель как взвешенная сумма

[Проекции](../../../math/linear-algebra/projections/) объяснили, какой вектор прогнозов ближе к ответам по сумме квадратов, а [градиентный спуск](../../../math/optimization-methods/gradient-descent/) — как уменьшать цель последовательными шагами. Теперь связываем эти идеи с обучением линейной модели по признакам.

Для d признаков $x_1,\ldots,x_d$ линейная модель выдаёт $\hat y=b+\sum_j w_jx_j$. Веса w меняют вклад каждого признака, b — свободный член. Это обучаемые параметры: их значения находят по train. Настройки процедуры обучения, например силу штрафа в следующем уроке, называют гиперпараметрами и выбирают по validation или внутренней перекрёстной проверке. Линейность относится к коэффициентам: среди признаков могут быть заранее вычисленные x² или длина текста. Это не означает, что любая связь в исходных координатах — прямая.

Подбирают коэффициенты так, чтобы средняя квадратичная ошибка на train была мала. Для обучения и проверки используют разные объекты. Малый обучающий штраф может означать удачное описание, но также утечку или слишком гибкий набор признаков.

**Функция потерь** задаёт, за какие ошибки платит обучение; **метрика** измеряет результат для выбранной задачи. Они могут совпадать: здесь обучаемся по MSE и можем измерять MSE на validation. Важно, на каких объектах посчитано число. Позже классификатор сможет обучаться по вероятностной потере, а проверяться по доле верных меток — эти показатели отвечают на разные вопросы.

## Разобранный пример: один коэффициент

Пусть свободного члена нет, признаки [1,2], ответы [2,5]. Сумма квадратов ошибок $S(w)=(w-2)^2+(2w-5)^2$. Производная $2(w-2)+4(2w-5)=10w-24$. Ноль достигается при w=2.4. Предсказания [2.4,4.8], знаковые ошибки «прогноз − ответ» [0.4,−0.2], MSE=(0.16+0.04)/2=0.1. Остатки в принятом ранее смысле «ответ − прогноз» имеют противоположные знаки: [−0.4,0.2]. Вторая производная 10 положительна, значит найден минимум.

Для общей таблицы коэффициенты находят численными методами линейной алгебры. Не нужно самостоятельно вычислять обратную матрицу: вырожденные или почти зависимые признаки делают такое решение ненадёжным. LinearRegression — готовая модель наименьших квадратов; fit_intercept=False в примере фиксирует b=0. Свойство coef_ после fit хранит коэффициенты.

Коэффициент не равен причинному эффекту. Если два столбца почти повторяют друг друга, их отдельные веса могут сильно меняться, сохраняя почти те же прогнозы. Следующий урок добавляет штраф, который выбирает другой компромисс.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
model = LinearRegression(fit_intercept=False)
model.fit([[1.], [2.]], [2., 5.])
assert np.allclose(model.coef_, [2.4])
assert np.allclose(model.predict([[1.], [2.]]), [2.4, 4.8])
print(model.coef_)
```
