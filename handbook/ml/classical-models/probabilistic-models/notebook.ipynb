## Два направления вопроса

Вероятностная модель описывает, какие данные ожидаются при разных классах. Априорная вероятность P(y) задаёт исходную долю класса, правдоподобие P(x|y) — вероятность признака при этом классе. Для классификации нужна обратная вероятность P(y|x), которую получают по Байесу: произведения P(y)P(x|y) нормируют суммой по классам.

Правдоподобие непрерывного признака может быть плотностью, а не вероятностью одной точки. Поэтому нельзя напрямую сравнивать значения моделей, использующих разные определения наблюдаемого события или единицы, без понимания формулы.

## Разобранный пример: два класса сообщений

Пусть 20% сообщений нежелательные. Слово «приз» встречается в 60% нежелательных и 5% обычных. Совместные веса равны 0.2·0.6=0.12 и 0.8·0.05=0.04. Среди сообщений со словом вероятность нежелательного класса 0.12/(0.12+0.04)=0.75. Нельзя сообщить 0.6: это вероятность слова при классе, а не класса при слове.

Если исходная доля нежелательных сообщений упадёт до 1%, те же правдоподобия дадут 0.006/(0.006+0.0495)≈0.108. Смена исходной частоты существенно меняет вывод даже при неизменной связи слова и класса.

## Оценивание модели

Вероятности признаков обычно неизвестны и оцениваются по train. Нулевой счётчик может давать слишком сильный вывод, будто новый вариант невозможен. Сглаживание добавляет заранее выбранную небольшую массу возможным значениям. Это предположение, а не обнаружение новых наблюдений.

Для нескольких признаков нужно описать совместное распределение. Следующий урок вводит сильное упрощение — условную независимость. До него не перемножайте вероятности отдельных слов без обоснования.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
priors = {"spam": 0.2, "regular": 0.8}
likelihood = {"spam": 0.6, "regular": 0.05}
weights = {label: priors[label] * likelihood[label] for label in priors}
total = sum(weights.values())
assert total > 0
posterior = {label: weight / total for label, weight in weights.items()}
assert abs(posterior["spam"] - 0.75) < 1e-12
print(posterior)
```
