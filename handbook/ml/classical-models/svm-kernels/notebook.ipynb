## Граница между классами

Для меток −1 и +1 линейный классификатор использует знак z=w·x+b. Граница z=0 называется гиперплоскостью: на плоскости признаков это прямая. Одну и ту же границу можно записать с разными множителями w,b, поэтому величина счёта не равна геометрическому расстоянию без нормировки.

Расстояние до границы равно |w·x+b|/||w|| при ненулевом w. SVM предпочитает границу с большим **зазором**, то есть запасом до ближайших важных обучающих объектов. Объекты, влияющие на положение границы, называют опорными.

## Разобранный пример: граница x=0

В одном измерении точки −2,−1 имеют класс −1, точки 1,2 — +1. Граница в нуле разделяет их с минимальным расстоянием 1. Запись w=1,b=0 даёт ближайшим точкам счёты −1,+1. Запись w=10,b=0 даёт ту же границу, а расстояния остаются прежними после деления на |w|.

Для шумных данных нельзя всегда требовать идеального разделения. **Мягкий зазор** допускает нарушения за штраф. Hinge-потеря max(0,1−yz) равна нулю, когда правильный знак имеет запас не меньше 1. Больший C в sklearn усиливает цену нарушений относительно штрафа весов; это не гарантия большего качества.

Признаки нужно масштабировать внутри обучения: геометрия зависит от единиц. SVC(kernel='linear') строит линейный SVM; его обычные счёты decision_function не являются вероятностями. Следующий урок заменит скалярное произведение ядром, сохраняя идею сравнения объектов.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.svm import SVC
model = SVC(kernel="linear", C=1.).fit([[-2.], [-1.], [1.], [2.]], [-1, -1, 1, 1])
assert model.predict([[-3.], [3.]]).tolist() == [-1, 1]
scores = model.decision_function([[-3.], [3.]])
assert scores[0] < 0 < scores[1]
print(scores, model.support_)  # support_ — индексы опорных объектов train
```
