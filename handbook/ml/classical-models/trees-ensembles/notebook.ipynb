## Последовательность вопросов

Дерево решений проверяет условия вроде «размах ≤15?». Внутренний узел задаёт вопрос, ребро выбирает ответ, лист выдаёт прогноз. Для числового признака узел сравнивает с порогом; разные ветви могут использовать разные признаки. В классификации лист обычно хранит частоты классов, в регрессии — средний числовой ответ.

Для выбора вопроса измеряют **нечистоту** группы. Индекс Джини равен $G=1-\sum_k p_k^2$. У чистой группы G=0, у равной смеси двух классов G=0.5. Разбиение оценивают средним G дочерних групп с весами их размеров. Максимальное уменьшение предпочтительнее в жадном шаге.

## Разобранный пример: идеальный вопрос

Есть четыре объекта с одним признаком [1,2,3,4] и метками [0,0,1,1]. До разбиения G=0.5. Порог 2.5 создаёт две чистые группы, средняя нечистота 0, улучшение 0.5. Порог 1.5 оставляет справа метки [0,1,1] с G=4/9; средняя нечистота 1/4·0+3/4·4/9=1/3. Первый вопрос лучше по этому критерию.

Жадный выбор смотрит на один следующий шаг и не гарантирует лучшее дерево среди всех возможных структур. Рост глубины способен выделять всё меньшие группы вплоть до отдельных шумных объектов. Поэтому отдельный критерий разбиения не заменяет ограничений и проверки.

DecisionTreeClassifier(max_depth=1) строит дерево только с одним уровнем вопросов. Параметр random_state фиксирует источники случайного выбора при равенствах. В примере predict показывает метки, выданные листьями; интерпретировать дерево можно без огромного рисунка, проследив один вопрос вручную.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(max_depth=1, random_state=17)
model.fit([[1.], [2.], [3.], [4.]], [0, 0, 1, 1])
assert model.predict([[1.5], [3.5]]).tolist() == [0, 1]
assert model.get_depth() == 1
print(model.predict_proba([[2.], [3.]]))
```
