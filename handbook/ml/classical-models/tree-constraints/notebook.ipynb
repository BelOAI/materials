## Чему научимся

Выбирать ограничения дерева по внутренней проверке и объяснять ненадёжность маленьких листьев.

## Почему чистый лист не всегда хорош

Лист с одним положительным объектом имеет чистоту 100%, но мало сведений о новых объектах. Глубокое дерево может запомнить случайные особенности train. Оно уменьшает обучающую ошибку, разделяя всё более мелкие группы, и становится чувствительным к небольшим изменениям данных.

max_depth ограничивает число уровней вопросов. min_samples_leaf требует минимальное число обучающих объектов в каждом листе. Эти настройки управляют разными свойствами: мелкое дерево ограничено глобально, минимальный лист запрещает слишком маленькие локальные решения.

## Разобранный пример: необычная точка

Признаки [0,1,2,3,4,5], метки [0,0,0,1,1,0]. Последний ноль может быть реальным исключением или шумом. Неограниченное дерево способно отделить его отдельным листом. Дерево с min_samples_leaf=2 не сможет создать одиночный лист. Оно намеренно допускает обучающую ошибку ради устойчивости.

Нельзя по виду последнего объекта решить, какое дерево лучше для будущего: нужны дополнительные данные или validation. Удаление исключения по проверочной метке также меняет задачу и может скрыть важную группу.

## Подбор и интерпретация

Сравнивайте настройки на одной схеме проверки и одной метрике. После выбора посмотрите глубину, число листьев и ошибки по группам. Ограничение сложности не защищает от утечки признаков: даже дерево глубины 1 прекрасно «предсказывает» ответ, если он спрятан в доступном столбце.

В ноутбуке показываем, что ограничение числа объектов листа действительно соблюдено. Свойство apply возвращает номер листа для каждого объекта, но эти номера — внутренние идентификаторы, а не классы. predict выдаёт метки.

`Counter` из стандартного модуля `collections` — словарь счётчиков: он считает, сколько раз встречается каждый номер листа. `get_depth()` сообщает фактическую глубину дерева.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
from collections import Counter
from sklearn.tree import DecisionTreeClassifier
X = [[float(i)] for i in range(6)]
y = [0, 0, 0, 1, 1, 0]
model = DecisionTreeClassifier(min_samples_leaf=2, random_state=17).fit(X, y)
leaf_counts = Counter(model.apply(X))
assert min(leaf_counts.values()) >= 2
print("Глубина:", model.get_depth(), "Размеры листьев:", sorted(leaf_counts.values()))
```
