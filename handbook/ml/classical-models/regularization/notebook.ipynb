## Зачем менять минимум ошибки

Если много похожих признаков, обучение может подобрать большие взаимно компенсирующие веса. Добавочный штраф предпочитает более простое решение. **Ridge** штрафует сумму квадратов коэффициентов, **Lasso** — сумму их модулей. Конкретные множители перед ошибкой и штрафом зависят от соглашения библиотеки, поэтому численное значение силы нельзя бездумно переносить между реализациями.

Штраф L2 гладко уменьшает веса, но обычно не делает их точно нулевыми. У L1 есть излом в нуле; для некоторых задач минимум возникает ровно там, и Lasso выполняет отбор признаков. Нулевой вес не доказывает бесполезность признака вообще: коррелирующие столбцы могут заменять друг друга.

## Разобранный пример: чаша со штрафом

Пусть λ≥0 — сила штрафа. Для учебной цели $(w-3)^2+\lambda w^2$ производная равна 2(w−3)+2λw. Минимум w=3/(1+λ). При λ=0 получаем 3, при λ=2 — 1. Штраф уменьшил вес ценой роста обучающей ошибки. Для цели $(w-3)^2+\lambda|w|$ при w>0 производная 2(w−3)+λ; при λ≥6 оптимум достигается в нуле.

## Единицы имеют значение

Если признак увеличить в сто раз, тот же прогноз можно сохранить, уменьшив вес в сто раз. Его штраф станет другим, хотя смысл измерения не изменился. Поэтому числовые признаки обычно масштабируют внутри конвейера перед регуляризованной моделью.

Параметр alpha задаёт силу штрафа в sklearn.Ridge и Lasso, но определения их целей не полностью одинаковы. Свободный член обычно не штрафуется. Силу выбирают по внутренней проверке; итоговый тест не используют для подбора alpha. В ноутбуке сравниваем коэффициенты одной реализации, не приравнивая её alpha к λ ручного примера.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
X = [[0.], [1.], [2.], [3.]]
y = [0., 2., 4., 6.]
weights = []
for alpha in [0.1, 10.]:
    model = Pipeline([("scale", StandardScaler()), ("ridge", Ridge(alpha=alpha))])
    model.fit(X, y)
    weights.append(model.named_steps["ridge"].coef_[0])
assert 0 < weights[1] < weights[0]
print(weights)
```

## Типичные ошибки

Не смешивайте разные нормировки целей. Не масштабируйте тест отдельно. Большее alpha не обязано улучшать новые прогнозы.
