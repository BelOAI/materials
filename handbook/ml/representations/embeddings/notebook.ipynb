## Координаты — способ описания

Эмбеддинг — векторное представление объекта, выбранное или выученное для определённой задачи. Координаты не обязаны иметь простые названия вроде массы и длины. Близость зависит от метода обучения, нормировки и выбранной метрики; похожие векторы не доказывают тождество или одинаковый правильный ответ.

Для сравнения направлений используют косинус $x\cdot z/(||x||||z||)$ при ненулевых векторах. Он не учитывает общий положительный масштаб, но всё ещё зависит от того, что закодировано. Для нулевого вектора формула не определена.

## Разобранный пример: потерянная координата

Объекты (0,0,1) и (0,0,−1) различаются на расстояние 2. Проекция на первые две координаты превращает оба в (0,0). Совпадение точек на рисунке не означает, что исходные объекты близки. Проекция потеряла именно различающую координату.

При визуализации многомерных данных полезно сравнивать соседей до и после преобразования, отмечать группы и смотреть несколько настроек. Красивые островки могут зависеть от параметров, случайного старта и отбора объектов. Они не заменяют оценку задачи на исходных данных.

## Как проверять представление

Задайте вопрос: требуется поиск похожих, классификация или сжатие? Для поиска можно проверять, попадают ли нужные объекты в верхние k соседей на независимой разметке. Для классификации учат простую модель поверх представления по честному разбиению. Если эмбеддинг обучался с ответами, его обучение также должно оставаться внутри train.

В ноутбуке векторы заданы автором вручную. Это иллюстрация свойств метрики и потери измерения, а не выученная модель языка или изображения.

`np.linalg.norm` вычисляет евклидову длину вектора. Проверка до и после удаления координаты показывает численную потерю расстояния.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
X = np.array([[0., 0., 1.], [0., 0., -1.]])
projected = X[:, :2]
assert np.linalg.norm(X[0] - X[1]) == 2
assert np.linalg.norm(projected[0] - projected[1]) == 0
x, z = np.array([1., 0.]), np.array([2., 0.])
cosine = (x @ z) / (np.linalg.norm(x) * np.linalg.norm(z))
assert cosine == 1
print("Исходное расстояние: 2; на плоской проекции: 0")
```
