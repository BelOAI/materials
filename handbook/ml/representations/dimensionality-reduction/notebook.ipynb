## Меняем координаты

В [SVD](../../../math/linear-algebra/svd/) мы сохраняли наиболее сильные направления матрицы. В PCA применим это к центрированной таблице признаков и получим новые координаты объектов. Это не новый способ разбиения на группы: задача здесь — сжать описание каждого объекта.

PCA, метод главных компонент, ищет перпендикулярные направления наибольшего разброса центрированных обучающих данных. Первая компонента сохраняет максимальную возможную дисперсию проекции на одну ось, следующая — максимальную оставшуюся при перпендикулярности предыдущим.

Это связано с SVD центрированной таблицы. Перед PCA выбирают масштаб признаков: сантиметры и миллиметры могут изменить направление наибольшего разброса. Метод не использует правильные ответы и не обещает сохранять наиболее полезную для классификации информацию.

## Разобранный пример: точки на прямой

Точки (−2,−2), (−1,−1), (1,1), (2,2) имеют нулевое среднее и лежат на направлении (1,1). Первая единичная ось u=(1,1)/√2. Координаты проекций −2√2,−√2,√2,2√2. Перпендикулярная компонента у всех ноль, поэтому одна компонента восстанавливает таблицу без ошибки.

Знак u произволен: ось −u описывает ту же прямую с противоположным знаком координат. Поэтому тест не должен требовать определённого знака компоненты, если проверяется геометрия.

## Сколько оставить

explained_variance_ratio_ сообщает доли обучающей дисперсии компонент. Большая накопленная доля полезна для сжатия, но проверка конечной задачи нужна отдельно. inverse_transform восстанавливает приближение исходных признаков; при усечении часть информации потеряна.

PCA обучают только на train; проверку преобразуют сохранёнными средним и осями. Повторное обучение PCA на тесте даёт другие координаты, с которыми уже обученная модель может быть несовместима.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.decomposition import PCA
X = np.array([[-2., -2.], [-1., -1.], [1., 1.], [2., 2.]])
model = PCA(n_components=1).fit(X)
Z = model.transform(X)
assert Z.shape == (4, 1)
assert np.allclose(model.inverse_transform(Z), X)
assert np.isclose(model.explained_variance_ratio_[0], 1.)
print(Z, model.explained_variance_ratio_)
```

## Типичные ошибки

Не забывайте центрирование и выбор масштаба. Не фиксируйте знак компоненты в тесте вместо проверки реконструкции.
