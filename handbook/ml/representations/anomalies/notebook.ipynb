## Ненормальное относительно чего

Аномалия — объект, который необычен относительно выбранного описания и модели. Большая масса может быть аномальной среди бумажных планеров и обычной среди игрушечных самолётов. Поэтому сначала задают сравнимую группу, единицы и момент измерения.

Аномальность не равна вредности или неверности. Редкий допустимый режим и опечатка могут выглядеть одинаково. Если есть готовые метки нарушений, задачу можно оценивать как классификацию; без них приходится использовать устойчивость, проверку специалистом и стоимость ручного просмотра.

## Разобранный пример: расстояние от центра

Для обучающих значений [8,9,10,11,12] медиана 10. Оценка |x−10| даёт объекту 11 значение 1, объекту 40 — 30. Порог 5 направит 40 на проверку. Это правило не доказывает, что 40 ошибочно, и не учитывает форму многомерного распределения.

Порог определяют на данных разработки или исходя из бюджета проверки. Отбор ровно k объектов с наибольшими оценками задаёт объём проверки, но не число настоящих проблем. Например, k=10 из 1000 — это 1% объектов; даже если все они нормальны, правило выделит десять. Для маленькой таблицы нужно заранее решить, как округлять процент до целого числа, а при равных оценках — как выбирать между объектами. Порог по 99%-квантили и отбор ровно 1% — разные правила: при одинаковых оценках порог может выделить другое число строк.

## Оценка без готовой истины

Смотрите, остаются ли объекты необычными при небольшом изменении train и разумном масштабе. Исследуйте ошибки по группам: новый пользователь может быть необычным просто из-за отсутствия истории. Ручная проверка требует независимого правила разметки, иначе специалист начнёт подтверждать всё, что подсказал алгоритм.

В ноутбуке используем прозрачный одномерный показатель, а не объявляем сложный детектор универсальным решением. Все данные авторские; порог фиксируется до просмотра нового значения.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
train = np.array([8., 9., 10., 11., 12.])
center = np.median(train)
new = np.array([11., 40.])
scores = np.abs(new - center)
flagged = scores > 5
assert scores.tolist() == [1., 30.]
assert flagged.tolist() == [False, True]
print(scores, flagged)
```
