## Не всякая группа похожа на шар

K-means тянет объекты к центрам и плохо описывает некоторые вытянутые или изогнутые группы. **DBSCAN** связывает области плотности: eps задаёт радиус соседства, min_samples — минимальное число объектов в нём с учётом самой точки. Плотные опорные области соединяются, граничные объекты могут присоединяться, а не охваченные группы точки отмечаются шумом, обычно меткой −1.

Точка называется **основной**, если в её eps-окрестности не меньше min_samples точек, включая её саму. Цепочки соседних основных точек образуют плотные группы. **Граничная** точка сама не основная, но находится рядом с основной и присоединяется к её группе. Остальные получают метку шума. Цепочка только из граничных точек не соединяет две группы; у границы нескольких групп результат может зависеть от порядка обработки.

Здесь шум — термин алгоритма, не доказательство ошибочности данных. Маленький отдельный настоящий класс тоже может получить −1. Одинаковый радиус плохо подходит группам с сильно разной плотностью.

## Разобранный пример: два тесных набора

На прямой точки [0,0.1,0.2] и [3,3.1,3.2], отдельно 10. При eps=0.25 и min_samples=2 внутри каждой тройки есть соседи, группы разделены, 10 изолирована. После перевода координат в другие единицы тот же численный eps означает иной радиус: масштаб — часть настройки.

## Иерархическая кластеризация

Агломеративный метод начинает с отдельных точек и постепенно объединяет ближайшие группы. **Связь**, linkage, определяет расстояние между группами: минимум парных расстояний, максимум, среднее или увеличение внутригрупповой суммы квадратов. Разные правила дают разные деревья объединений. Срез дерева на нужной высоте или числе групп задаёт разбиение.

Например, для групп [0,1] и [3] парные расстояния равны 3 и 2. Минимальная связь даёт 2, максимальная (complete) — 3, средняя — 2.5. Значит, «ближайшие группы» имеет смысл только после выбора правила связи. Код ниже использует complete.

При минимальной связи цепочка близких друг к другу точек может соединить далёкие области. Поэтому рисунок дерева сам по себе не доказывает, что группы соответствуют естественным категориям. В ноутбуке сравниваем два метода на одном небольшом наборе, проверяя свойства групп, а не точные номера меток.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.cluster import DBSCAN, AgglomerativeClustering
X = np.array([[0.], [0.1], [0.2], [3.], [3.1], [3.2], [10.]])
labels = DBSCAN(eps=0.25, min_samples=2).fit_predict(X)
assert labels[-1] == -1
assert labels[0] == labels[1] == labels[2]
assert labels[3] == labels[4] == labels[5] != labels[0]
hierarchy = AgglomerativeClustering(n_clusters=3, linkage="complete").fit_predict(X)
assert len(set(hierarchy)) == 3
print(labels, hierarchy)
```
