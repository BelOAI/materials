## Группы без правильных меток

Кластеризация ищет структуру признаков без готового правильного класса. **K-means** заранее получает число K и минимизирует сумму квадратов расстояний объектов до назначенных центров. Кластер не обязательно совпадает с предметной категорией: алгоритм видит выбранную геометрию, а не смысл названий.

Алгоритм чередует два действия: назначает объект ближайшему центру, затем заменяет центр средним его объектов. При фиксированных назначениях среднее минимизирует сумму квадратов; при фиксированных центрах ближайший выбор также её не увеличивает. Но общий минимум может зависеть от старта.

## Разобранный пример: одна прямая

Точки [0,2,8,10], стартовые центры 0 и 8. Первые две точки относятся к первому центру, последние ко второму. Новые центры (0+2)/2=1 и (8+10)/2=9. Назначения не меняются; сумма квадратов равна 1+1+1+1=4.

Если одна координата измерена тысячами, она может доминировать над другой. Масштабирование меняет задачу, поэтому его выбирают по смыслу. Для новых объектов используют сохранённые масштабы и центры, а не повторно обучают кластеры на проверке.

## Номера и выбор K

Номера кластеров произвольны: метки [0,0,1,1] и [1,1,0,0] могут описывать одно разбиение. С ростом K обучающая сумма квадратов обычно падает, поэтому нельзя выбрать K только по её минимуму. Нужны устойчивость, интерпретируемость и цель применения. В sklearn n_init задаёт число запусков с разными стартами; random_state фиксирует случайность. Малый пример не доказывает качество для сложных форм групп.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.cluster import KMeans
X = np.array([[0.], [2.], [8.], [10.]])
model = KMeans(n_clusters=2, n_init=10, random_state=17).fit(X)
assert np.allclose(sorted(model.cluster_centers_[:, 0]), [1., 9.])
assert np.isclose(model.inertia_, 4.)  # inertia_ — сумма квадратов до своих центров
assert model.predict([[1.]])[0] == model.predict([[0.]])[0]
print(model.labels_)
```
