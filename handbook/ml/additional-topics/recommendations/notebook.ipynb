Строить рекомендательный baseline и различать холодный старт пользователя и объекта.

## Предсказываем не только оценку

Рекомендательная задача выбирает объекты, полезные конкретному пользователю: книги, занятия или товары. Запись взаимодействия может означать просмотр, выбор или высокую оценку; эти события не тождественны. Отсутствие клика не всегда отрицательный ответ: пользователь мог не увидеть объект.

Baseline популярности предлагает самые частые объекты из прошлого. Персонализация дополнительно учитывает историю пользователя. Контентный метод сравнивает признаки объектов, коллаборативный — структуру взаимодействий пользователей и объектов.

## Разобранный пример: новый читатель

Книгу А раньше выбирали пять раз, Б — три, В — один. Новому читателю без истории можно предложить А,Б по популярности. Знакомому читателю, уже выбравшему А, исключим её из кандидатов и предложим Б,В. Правило исключения зависит от задачи: повторное прослушивание песни может быть уместным.

**Холодный старт пользователя** — недостаток его истории. Холодный старт объекта — отсутствие взаимодействий с новой книгой. Популярность нового объекта ноль, даже если он интересен; помогают содержательные признаки или отдельная стратегия исследования.

## Проверка рекомендаций

Разделение должно имитировать момент рекомендации. Будущие клики нельзя включать в популярность прошлого. Для новых пользователей полезно отдельное групповое разбиение. Список кандидатов также часть оценки: ранжировать десять известных хороших объектов легче, чем весь каталог.

В ноутбуке данные — авторские прошлые выборы, не реальные читатели. Counter считает частоты, sorted упорядочивает по убыванию частоты и имени при равенстве. Это прозрачный baseline без обещаний персонального качества.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
from collections import Counter
history = ["А"]*5 + ["Б"]*3 + ["В"]
counts = Counter(history)
ranked = [item for negative_count, item in sorted([(-count, item) for item, count in counts.items()])]
seen = {"А"}
recommended = [item for item in ranked if item not in seen]
assert ranked[:2] == ["А", "Б"]
assert recommended == ["Б", "В"]
print(recommended)
```
