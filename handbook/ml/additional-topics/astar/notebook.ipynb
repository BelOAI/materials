## Направляем поиск к цели

A* выбирает следующий узел по f(s)=g(s)+h(s). g — уже известная стоимость от старта, h — оценка оставшейся стоимости. Эвристика не является ценой пройденных рёбер и не должна незаметно изменять цель задачи.

**Допустимая** эвристика не завышает настоящую минимальную оставшуюся цену. **Согласованность** требует h(s)≤c(s,s′)+h(s′) для каждого перехода. Она помогает безопасно обрабатывать граф без неожиданных улучшений закрытых узлов. При одной лишь допустимости может потребоваться повторное открытие состояний.

## Разобранный пример: клетчатая карта

На прямоугольной сетке разрешены шаги вверх, вниз, влево, вправо цены 1. Манхэттенское расстояние |x−x_goal|+|y−y_goal| — нижняя оценка: каждый шаг меняет сумму не более чем на 1. Стены могут увеличить настоящий путь, но не делают эту оценку завышенной. При разрешённых дешёвых диагоналях прежняя оценка уже требует пересмотра.

Если h=0, A* превращается в поиск по стоимости. Если умножить допустимую h на 10, поиск может стать быстрее на некоторых примерах, но гарантия оптимального пути обычного A* теряется.

## Цена эвристики

Полезная оценка должна не только направлять поиск, но и быстро вычисляться. Слишком дорогой анализ каждого узла может съесть выигрыш. При равных f правило выбора влияет на число рассмотренных состояний, поэтому его фиксируют для повторяемости.

В ноутбуке проверяем манхэттенскую эвристику для соседних клеток. Это тест важного свойства, не полный автоматический доказатель для произвольной карты и действий.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
goal = (4, 6)
def h(point):
    return abs(point[0]-goal[0]) + abs(point[1]-goal[1])
assert h((1, 2)) == 7
assert h(goal) == 0
for x in range(4):
    for y in range(4):
        for dx, dy in [(1,0), (-1,0), (0,1), (0,-1)]:
            assert h((x,y)) <= 1 + h((x+dx,y+dy))
print("Проверены локальные неравенства эвристики")
```
