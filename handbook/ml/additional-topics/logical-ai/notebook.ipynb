## Истина, а не степень уверенности

Высказывание — утверждение, которому в выбранной модели приписывают истину или ложь. A∧B означает оба, A∨B — хотя бы одно, ¬A — отрицание. Эти операции знакомы по условиям, но теперь описывают знания и ограничения задачи.

Импликация A→B запрещает случай, когда A истинно, а B ложно. Если A ложно, правило само по себе не сообщает, истинно ли B. Поэтому «если дождь, дорожка мокрая» не означает «если дорожка мокрая, идёт дождь»: её могли полить.

## Разобранный пример: предметы и свойства

Предикат — утверждение с аргументом, например Ученик(Маша) или Посещает(Маша,Кружок). Правило «для каждого x, если Ученик(x) и Записан(x), то Допущен(x)» относится к объектам области. Знак ∀ читается «для всех», ∃ — «существует». Названия и область объектов задают до применения правил.

Если известны Ученик(Маша) и Записан(Маша), правило даёт Допущен(Маша). Отсутствие записи о Пете не доказывает ¬Допущен(Петя), если в системе нет отдельного соглашения о полноте фактов.

## Границы формализации

Логическое правило описывает заданные отношения, а не автоматически извлекает факты из реального мира. Ввод неверного факта может привести к формально правильному, но предметно ложному выводу. Вероятность 0.8 также не равна логической истине без отдельного правила принятия решения.

В ноутбуке строим таблицу истинности импликации через «не A или B». Это перевод её определения в знакомые логические операции, не универсальная система доказательства.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
def implies(a, b):
    return (not a) or b
truth_table = [(a, b, implies(a,b)) for a in [False, True] for b in [False, True]]
assert implies(True, False) is False
assert implies(False, False) is True
assert implies(False, True) is True
print(truth_table)
```
