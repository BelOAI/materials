## План как путь

Состояние описывает положение задачи, действие переводит его в другое состояние, цель проверяет завершение. Например, робот находится в клетке, действия — разрешённые переходы, цель — нужная клетка. Состояние должно включать всё, что влияет на доступные действия: если робот может нести ключ, одних координат может быть мало.

BFS обрабатывает состояния по числу шагов и находит путь с минимальным числом рёбер в невзвешенном графе. При разных неотрицательных ценах нужен поиск по накопленной стоимости: сначала извлекается состояние с наименьшей известной ценой. Это идея алгоритма Дейкстры, или uniform-cost search.

## Разобранный пример: меньше шагов, дороже

Есть прямое ребро А→Г цены 10 и путь А→Б→Г с ценами 2 и 3. BFS предпочёл бы один переход, а поиск по стоимости выберет сумму 5. После А фронтир содержит Б с ценой 2 и Г с ценой 10. Извлекая Б, улучшаем цену Г до 5.

Фронтир — найденные, но ещё не окончательно обработанные состояния. Нужно хранить лучшую известную цену и отбрасывать устаревшие записи очереди. При отрицательных весах рёбер описанный алгоритм может дать неверный ответ.

## Проверяем постановку

Граф может быть задан явно или порождаться действиями. Повторы состояний способны раздуть поиск без visited или таблицы цен. Восстановление пути требует хранить предшественника, а одна минимальная цена не сообщает последовательность действий.

В ноутбуке heapq — стандартная очередь с приоритетом; heappop извлекает минимальный кортеж по цене. Пары соседей содержат вершину и неотрицательную стоимость. При улучшении цены сохраняем предшественника в `previous`. После поиска идём от цели назад до старта; метод списка `reverse()` разворачивает накопленный путь на месте. В этом графе цель достижима; для недостижимой цели сначала проверили бы её наличие в `distance`.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import heapq
graph = {"А": [("Г", 10), ("Б", 2)], "Б": [("Г", 3)], "Г": []}
distance = {"А": 0}
previous = {"А": None}
frontier = [(0, "А")]
while frontier:
    cost, node = heapq.heappop(frontier)
    if cost != distance[node]:
        continue
    for neighbor, weight in graph[node]:
        assert weight >= 0
        candidate = cost + weight
        if candidate < distance.get(neighbor, float("inf")):
            distance[neighbor] = candidate
            previous[neighbor] = node
            heapq.heappush(frontier, (candidate, neighbor))
assert distance["Г"] == 5
path = []
node = "Г"
while node is not None:
    path.append(node)
    node = previous[node]
path.reverse()
assert path == ["А", "Б", "Г"]
print(distance, path)
```
