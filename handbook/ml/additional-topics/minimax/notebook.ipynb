## Другой участник тоже выбирает

В игре с полной информацией и противоположными целями один игрок максимизирует итоговую оценку, другой минимизирует. **Minimax** разворачивает дерево действий и передаёт оценки от листьев вверх: на ходе первого берётся максимум, второго — минимум. Это модель рационального противника, а не случайного выбора.

Терминальные оценки задают исход игры. Если дерево обрезано раньше конца, вместо них нужна эвристическая оценка. Ошибка этой оценки способна изменить выбор даже при безупречном вычислении minimax.

## Разобранный пример: два решения

Из корня игрок MAX выбирает ветку А или Б. В А соперник MIN выбирает итог 3 или 5, значит ценность А равна 3. В Б он выбирает 2 или 9, значит ценность Б равна 2. MAX выбирает А, хотя самый большой лист 9 лежит в Б. Рассчитывать, что соперник обязательно позволит 9, неверно в принятой модели.

Для случайной среды вместо минимума часто нужно ожидание по вероятностям. Замена противника случайным событием — другая постановка. Планирование также требует модели последствий действий; обучение с подкреплением может учить ценность без полной известной модели.

## Ограничения поиска

При ветвлении b и глубине d дерево содержит порядка b^d листьев. Поэтому глубина и время ограничены. Одинаковые состояния могут встречаться разными путями; при повторном использовании оценки нужно учитывать, чей сейчас ход, и другие важные сведения о состоянии.

В ноутбуке маленькое дерево задано явно, числа — условный выигрыш MAX. Мы проверяем выбор, а не обучаем стратегию по данным.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
branches = {"А": [3., 5.], "Б": [2., 9.]}
values = {name: min(leaves) for name, leaves in branches.items()}
# Заранее заданное правило равенства: первая по алфавиту ветка.
ranked = sorted([(-value, name) for name, value in values.items()])
chosen = ranked[0][1]
assert values == {"А": 3., "Б": 2.}
assert chosen == "А"
print(values, chosen)
```
