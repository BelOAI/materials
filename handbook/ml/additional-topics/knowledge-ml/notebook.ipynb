## Два источника решения

Статистическая модель получает закономерности из примеров и может ошибаться на новом объекте. Символическая система применяет явно заданные факты, отношения и ограничения. Она может точно выполнить правило, но не проверить, верно ли введённое наблюдение.

База знаний хранит предметные сущности и связи: например, занятие проводится в кабинете, кабинет имеет вместимость, ученик записан на занятие. Названия, идентификаторы и смысл отношений должны быть согласованы. Два имени одного кабинета создают ложное впечатление разных ресурсов.

## Разобранный пример: прогноз и запрет

Модель оценивает вероятность интереса к кружку. Правило запрещает рекомендовать занятие, если в нём нет мест или оно пересекается с обязательным уроком. Высокий прогноз интереса не отменяет ограничение. При этом неверные сведения о расписании могут запретить хороший вариант: правила не исправляют качество исходных данных.

Гибридное решение явно разделяет этапы: модель ранжирует допустимые интересы, правила проверяют ограничения, система возвращает рекомендацию или отказ. Порядок этапов влияет на вычисления, но ограничения должны соблюдаться в конечном ответе.

## Что проверять отдельно

Оценивайте прогноз, полноту и непротиворечивость знаний, соблюдение правил и случаи отказа. Нельзя назвать всё решение объяснимым только потому, что последний шаг содержит if. Пользователю нужно понимать, какие факты использованы, насколько они надёжны и кто может исправить ошибку.

В ноутбуке вероятностные оценки заданы вручную. Программа только фильтрует авторские варианты по доступности и затем сортирует по оценке. Здесь нет обученной модели распознавания интересов или автоматического извлечения фактов.

В проверках `all(список_условий)` возвращает True, только если истинны все условия списка. Это позволяет одним утверждением проверить каждый параметр или каждую строку результата. Для пустого списка `all` тоже даёт True; поэтому непустоту данных при необходимости проверяют отдельно.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
candidates = [
    {"name": "А", "score": 0.9, "places": 0, "conflict": False},
    {"name": "Б", "score": 0.8, "places": 2, "conflict": False},
    {"name": "В", "score": 0.7, "places": 4, "conflict": True},
]
allowed = [row for row in candidates if row["places"] > 0 and not row["conflict"]]
ranked = sorted([(-row["score"], row["name"]) for row in allowed])
assert [name for _, name in ranked] == ["Б"]
assert all([row["places"] > 0 and not row["conflict"] for row in allowed])
print(ranked)
```
