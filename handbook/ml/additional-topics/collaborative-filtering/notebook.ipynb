## Пользователи и объекты

Матрица взаимодействий имеет пользователей по строкам и объекты по столбцам. Ячейка может хранить оценку или число событий. Неизвестная оценка отличается от известного нуля; выбор значения пропусков меняет задачу.

Коллаборативный подход использует сходство историй. Два человека, выбравшие похожие объекты, могут иметь похожие следующие предпочтения, но это предположение проверяют, а не считают законом. Пользовательская и объектная похожесть требуют достаточного пересечения наблюдений.

## Разобранный пример: скрытые факторы

При факторизации пользователю назначают вектор u, объекту v, прогнозируют u·v, иногда добавляя общий и персональные сдвиги. Для u=(1,2), v=(3,0.5) прогноз 3+1=4. Факторы обучаются по наблюдаемым взаимодействиям с регуляризацией. Их координаты не обязаны иметь названия вроде «любит фантастику».

Матрица ранга r описывается примерно r(n_users+n_items) числами вместо полного произведения размеров, но алгоритму всё ещё нужны данные и вычисления. Малый ранг может потерять индивидуальные предпочтения. Новому пользователю или объекту без наблюдений нечего надёжно подбирать: холодный старт остаётся.

## Честность оценки

Не обучайте факторы на будущих тестовых взаимодействиях. Для рекомендаций мало точно восстанавливать известные оценки: важно, какие объекты окажутся в начале списка. Нужны кандидаты и ранжирующая метрика.

В ноутбуке факторы заданы вручную: мы проверяем произведение и формы, не выдаём их за выученные предпочтения настоящих людей. Непрерывный прогноз также не обязан автоматически лежать в шкале допустимых оценок.

`V.T` меняет оси таблицы факторов объектов. Поэтому каждая клетка `U @ V.T` — скалярное произведение одного пользователя и одного объекта.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
U = np.array([[1., 2.], [0., 1.]])
V = np.array([[3., 0.5], [1., 1.], [0., 2.]])
scores = U @ V.T
assert scores.shape == (2, 3)
assert scores[0, 0] == 4
assert np.allclose(scores[1], [0.5, 1., 2.])
print(scores)
```
