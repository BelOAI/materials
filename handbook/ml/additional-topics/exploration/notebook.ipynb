## Неизвестное действие может быть лучше

Если всегда выбирать текущее самое выгодное действие, агент может не узнать о другом, более полезном. **Exploration**, разведка, пробует неизвестные варианты; **exploitation**, использование, выбирает лучшую текущую оценку. Компромисс зависит от цены опыта и времени.

ε-greedy выбирает случайное действие с вероятностью ε, иначе одно из действий с максимальной Q. При ε=0 исследования нет, при ε=1 выбор полностью случайный. Правило разрешения равных оценок задают отдельно; постоянный выбор первого может создавать перекос.

## Разобранный пример: два рычага

У действия А текущая оценка 0.6, у Б 0.5. При ε=0.2 и случайном выборе поровну шанс Б равен 0.2·0.5=0.1. Шанс А 0.8+0.1=0.9. Это не 0.8 и 0.2: случайная ветка тоже иногда выбирает А.

## Оценка отдельно от обучения

Для сравнения фиксируют политику и проводят новые эпизоды с заранее заданными стартами. Сообщают средний возврат и разброс, не выбирая лучший эпизод. Исследующая политика обучения и детерминированная политика применения могут иметь разные результаты; нужно написать, какую оценивали.

Сравнение на старых данных другой политики требует дополнительных предположений о покрытии действий. Нельзя уверенно судить о действии, которое в журнале никогда не пробовали. В реальных задачах случайное исследование также ограничено допустимостью и ценой ошибок.

В ноутбуке выбираем действия из заданной таблицы с локальным генератором. Это проверка вероятностей и повторяемости, не полноценное обучение стратегии.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
def actions(seed, count=10):
    rng = np.random.default_rng(seed)
    result = []
    for _ in range(count):
        if rng.random() < 0.2:
            result.append(int(rng.integers(0, 2)))
        else:
            result.append(0)
    return result
assert actions(17) == actions(17)
assert set(actions(17)).issubset({0, 1})
assert np.isclose(0.2 / 2, 0.1)
print(actions(17))
```
