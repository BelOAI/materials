## Чему научимся

Различать состояние, действие, награду и возврат в задаче взаимодействия со средой.

## Ответ влияет на следующий опыт

В обучении с подкреплением агент выбирает действия, среда возвращает награду и новое состояние. В отличие от обычной классификации, агент может менять будущие данные своими решениями. Правильное действие не всегда дано готовой меткой.

Состояние должно содержать достаточно сведений для выбранной модели переходов. Марковское предположение означает, что распределение следующего состояния и награды при действии зависит от текущего состояния, а не дополнительно от всей истории. Это свойство постановки, которое может нарушаться при неполных наблюдениях.

## Разобранный пример: короткий маршрут

Награды по трём действиям [0,0,10]. **Возврат** — суммарная будущая награда, часто с дисконтированием: G=r₀+γr₁+γ²r₂, где 0≤γ≤1. При γ=0.9 получаем 8.1. Немедленная награда первого шага ноль, но действие может быть полезно ради последующего выигрыша.

Политика, или стратегия, задаёт выбор действия по состоянию: детерминированный либо случайный. Эпизод — один законченный опыт от старта до завершения. Для бесконечного продолжения при ограниченных по модулю наградах условие 0≤γ<1 делает сумму возврата конечной. При γ=1 такой гарантии нет: бесконечная последовательность наград 1 даёт неограниченную сумму. В наших конечных эпизодах γ=1 допустимо.

## Задаём среду честно

Нужно указать допустимые действия, переходы, завершение и правило награды. Если награда поощряет число нажатий, агент может научиться нажимать без достижения полезной цели. Высокая награда подтверждает оптимизацию формального критерия, а не правильность самого критерия.

В ноутбуке среда — авторский численный эпизод без внешнего сервиса. Считаем возврат известной последовательности; это не обучение универсального агента.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
rewards = [0., 0., 10.]
gamma = 0.9
value = sum([(gamma ** t) * reward for t, reward in enumerate(rewards)])
assert abs(value - 8.1) < 1e-12
assert 0 <= gamma <= 1
print("Возврат известного эпизода:", value)
```
