## Какую величину и насколько вперёд

Временной ряд — наблюдения величины, упорядоченные по времени. Горизонт прогноза показывает, на сколько шагов вперёд нужен ответ. **Лаг** — прошлое значение, например yₜ₋₁. Тренд — длительное изменение уровня, сезонность — повторяющийся рисунок с периодом.

Наивный baseline предсказывает последнее известное значение. Сезонный baseline использует аналогичную позицию прошлого периода. Они часто сильнее среднего для меняющегося ряда, но требуют подходящей частоты и доступной истории.

## Разобранный пример: следующий день

Известны значения [10,12,11,15]. Перед пятым днём лаг 1 равен 15, среднее последних двух известных равно 13. Нельзя включать значение пятого дня в его же среднее. Если задача выдаёт прогноз сразу на неделю, истинные ответы первых дней этой недели также недоступны при построении дальних лагов.

Скользящая проверка обучается на прошлом, проверяется на следующем отрезке, затем граница движется вперёд. Размер горизонта и правило обновления модели должны совпадать с будущим применением. Случайное разбиение строк теряет это направление.

## Пропуски и календарь

Проверьте регулярность времени: lag 1 по строкам может означать вчера или неделю назад при пропущенных датах. День недели полезен, если календарь действительно связан с явлением. Агрегаты должны быть сдвинуты так, чтобы не включать текущий неизвестный ответ.

В ноутбуке строим одноступенчатый baseline и явные пары прошлого и следующего значения. Это не независимая оценка долгого горизонта.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
series = np.array([10., 12., 11., 15., 14., 18.])
lag_one = series[:-1]
next_values = series[1:]
assert lag_one.shape == next_values.shape
assert lag_one.tolist() == [10., 12., 11., 15., 14.]
assert (series[2:4].mean()) == 13
print("Одношаговые ошибки baseline:", lag_one-next_values)
```
