## Два этапа большого каталога

Retrieval, поиск кандидатов, быстро выбирает небольшой набор из большого каталога. Ранжирование затем упорядочивает этих кандидатов более подробной моделью. Если нужный объект не попал в кандидаты, ранжировщик уже не сможет поставить его первым.

Метрики верхних позиций учитывают, что пользователь видит ограниченную часть списка. Precision@k — доля релевантных среди k предложенных, recall@k — доля найденных среди всех известных релевантных для этого случая. Список должен содержать разные объекты; повтор одного полезного объекта не создаёт новых попаданий.

## Разобранный пример: два попадания

Правильные полезные объекты {А,В,Г}, top-3=[А,Б,В]. Попаданий два. Precision@3=2/3, recall@3=2/3. Если релевантных было четыре, precision осталась бы 2/3, recall стала бы 1/2. Поэтому знаменатели нельзя путать.

Для ранга первого полезного объекта используют reciprocal rank — обратную позицию. Если первое попадание на третьем месте, значение 1/3. При отсутствии попаданий по заданному ограничению обычно ставят 0, явно описывая протокол.

## Что неизвестно в разметке

В рекомендациях часто известна лишь часть полезных объектов. Тогда метрика относится к выбранной процедуре наблюдения, а не всей истинной полезности каталога. Укажите временной горизонт, кандидатов, увиденные объекты и правила отбора отрицательных примеров.

В ноутбуке релевантность задана автором полностью, чтобы арифметика была проверяемой. Настоящую систему поиска этот короткий список не моделирует целиком.

`enumerate(ranked, 1)` нумерует позиции начиная с 1. Первый найденный релевантный объект задаёт reciprocal rank; при отсутствии оставляем 0.

Для этих формул k — положительное целое, не превышающее длину списка различных выданных объектов. При пустом списке или отсутствии релевантных объектов некоторые знаменатели равны нулю; заранее задайте соглашение и отдельно отметьте такие запросы. Нельзя молча делить число попаданий на произвольный k при списке другой длины.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
relevant = {"А", "В", "Г"}
ranked = ["А", "Б", "В"]
k = 3
if type(k) != int or not 1 <= k <= len(ranked):
    raise ValueError("Нужно целое k от 1 до длины выдачи")
if len(relevant) == 0:
    raise ValueError("Полнота не определена без релевантных объектов")
assert len(set(ranked)) == len(ranked)
hits = len(set(ranked[:k]) & relevant)
precision = hits / k
recall = hits / len(relevant)
assert precision == recall == 2 / 3
first = None
for i, item in enumerate(ranked, 1):
    if item in relevant:
        first = i
        break
rr = 0. if first is None else 1 / first
assert rr == 1
print(precision, recall, rr)
```
