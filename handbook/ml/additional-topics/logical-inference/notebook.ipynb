## Вывод шаг за шагом

Прямой вывод начинает с известных фактов и добавляет следствия правил, у которых все условия уже выполнены. Для конечного множества допустимых фактов повторяют проходы, пока ничего не меняется. Это **неподвижная точка** процедуры: следующий проход не добавляет знаний.

Цикл правил A→B и B→A не обязан зациклить корректный вывод, если факт добавляют только один раз. Однако он и не создаёт A или B из пустоты: нужна исходная посылка или другое правило.

## Разобранный пример: цепочка

Факт A, правила A→B, B→C. Первый проход может добавить B, затем C; если порядок правил обратный, C появится на следующем проходе. Итог {A,B,C} одинаков при завершении. Если остановиться после одного произвольного прохода, результат может зависеть от порядка записи.

**Выполнимость** набора условий означает существование хотя бы одной совместимой расстановки истин и ложей. Условия A∨B и ¬A выполнимы: A=False,B=True. Условия A и ¬A одновременно невыполнимы в обычной двузначной логике.

## Ограничения как задача

Планирование расписания можно описать запретами: один ученик не посещает два кабинета одновременно, число мест ограничено. Решение — назначение, удовлетворяющее всем условиям. Наличие решения не доказывает, что именно так устроен реальный мир; оно относится к модели ограничений.

В ноутбуке перебираем четыре назначения двух булевых переменных. Для большой задачи полный перебор растёт как 2^n и быстро становится дорогим. Поэтому маленький пример прозрачен, но не претендует на промышленный решатель.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
facts = {"A"}
rules = [("B", "C"), ("A", "B")]
while True:
    before = facts.copy()
    for condition, consequence in rules:
        if condition in facts:
            facts.add(consequence)
    if facts == before:
        break
assert facts == {"A", "B", "C"}
solutions = [(a,b) for a in [False,True] for b in [False,True] if (a or b) and not a]
assert solutions == [(False, True)]
print(facts, solutions)
```
