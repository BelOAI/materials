## Ценность состояния и действия

V(s) — ожидаемый возврат при старте из состояния s и следовании выбранной политике. Q(s,a) уточняет первое действие. Эти значения включают будущее и не совпадают с немедленной наградой.

Идея Беллмана разбивает будущее на ближайшую награду и оставшийся возврат. Для оптимального действия целевое выражение имеет вид r+γ maxₐ′Q(s′,a′). В терминальном состоянии продолжения нет, его добавка равна нулю. Максимум по допустимым действиям нельзя заменять случайным числом из другой строки таблицы.

Полное **уравнение Беллмана для оптимальной ценности** записывают как $Q^*(s,a)=E[r+\gamma\max_{a'}Q^*(s',a')\mid s,a]$. Звёздочка означает лучший достижимый ожидаемый возврат, штрих — следующее состояние или действие. Ожидание усредняет возможные награды и переходы среды при выбранных s,a. Например, если после действия эпизод завершается и награда равна 0 или 2 с равными шансами, ценность этого действия равна $0.5\cdot0+0.5\cdot2=1$, а не обязательно награде одного опыта. В Q-learning отдельное наблюдение $r+\gamma\max Q$ служит шумной целью обновления; это не готовое точное ожидание.

## Разобранный пример: корректировка оценки

Текущая Q(s,a)=2, наблюдали r=1,γ=0.9, лучший следующий Q=4. Цель 1+0.9·4=4.6. При размере обновления α=0.5 получаем Q_new=2+0.5·(4.6−2)=3.3. Это движение к новой цели, а не полная замена текущей таблицы.

**Q-learning** повторяет такие обновления по переходам. Он учится оценивать оптимальный выбор, даже если сбор данных использует исследующую политику. Сходимость требует условий. В конечной табличной задаче важно достаточно часто посещать пары «состояние, действие» и правильно выбирать шаги обучения. Один пример ещё не доказывает, что алгоритм научился выбирать действия.

## Таблица и масштаб

При n состояний и m действий храним примерно nm значений. Для изображений в качестве состояний такая таблица нереалистична; нужны другие представления. Здесь ограничиваемся маленькой учебной средой и явным терминальным переходом.

В ноутбуке словарь хранит пару «состояние, действие». Функция update отдельно принимает флаг terminal и не добавляет будущую ценность при завершении.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
def update(old, reward, next_best, alpha, gamma, terminal=False):
    target = reward if terminal else reward + gamma * next_best
    return old + alpha * (target - old)
assert abs(update(2., 1., 4., 0.5, 0.9) - 3.3) < 1e-12
assert update(1., 5., 999., 1., 0.9, terminal=True) == 5.
Q = {("start", "right"): 0.}
for _ in range(10):
    Q[("start", "right")] = update(Q[("start", "right")], 1., 0., 0.5, 0.9, terminal=True)
assert 0.99 < Q[("start", "right")] < 1
print(Q)
```

## Типичные ошибки
