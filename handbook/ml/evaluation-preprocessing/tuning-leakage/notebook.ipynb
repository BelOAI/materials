## Что выбирает человек

Коэффициенты линейной модели находят при обучении. Число соседей, глубина дерева, сила регуляризации задаются снаружи — это **гиперпараметры**. Их выбирают по внутренней проверке. После выбора фиксируют весь способ решения: признаки, подготовку, алгоритм и настройки.

Утечка — использование информации, которой не будет в момент реального предсказания или которая должна остаться скрытой проверке. Это может быть прямой ответ в столбце, среднее по всей таблице, другой снимок того же объекта или ручной выбор по тестовой оценке. Удаление столбца с названием target не гарантирует честности.

## Разобранный пример: таблица вариантов

На validation модель с k=1 получила 0.8, с k=3 — 0.9, с k=5 — 0.9. Правило равенства решили заранее: при равном качестве предпочитаем меньшее k. Значит, выбираем 3. Теперь можно обучить согласованную финальную модель и один раз измерить тест. Нельзя после результата теста выбрать 5 и продолжать называть тест независимым.

Лидерборд — оценка отправленного решения на скрытой части. Если много раз менять решение по этому числу, оно становится проверочной обратной связью, по которой тоже можно переобучиться. Полезны ограниченный план экспериментов и локальная схема проверки, соответствующая задаче.

## Проверяем происхождение столбца

Для каждого признака спросите: из каких записей он вычислен и когда они доступны? Например, «средняя оценка ученика за год» недоступна при прогнозе в сентябре. Конвейер защищает порядок обучаемых преобразований, но не исправляет признак из будущего.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
candidates = [{"k": 1, "validation": 0.8}, {"k": 3, "validation": 0.9}, {"k": 5, "validation": 0.9}]
# sorted упорядочивает кортежи сначала по минус-качеству, затем по k.
ranked = sorted([(-row["validation"], row["k"]) for row in candidates])
chosen_k = ranked[0][1]
assert chosen_k == 3
assert all(["test" not in row for row in candidates])
print("Зафиксированное число соседей:", chosen_k)
```
