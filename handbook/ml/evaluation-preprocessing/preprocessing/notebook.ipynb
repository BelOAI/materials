## Пропуск не равен нулю

В таблице масс [8,10,пропуск,12] неизвестная масса не стала нулевой. Ноль означал бы измеренную нулевую массу и изменил среднее. Сначала выясните причину: прибор не работал, поле неприменимо или значение скрыто по правилу. Сама причина отсутствия иногда несёт информацию, но должна быть доступна при прогнозировании.

Заполнение медианой берёт центральное значение обучающего ряда. Для [8,10,12] это 10. Преобразование учится на train и затем без пересчёта применяется к validation и test. Можно добавить отдельный индикатор «значение отсутствовало», чтобы модель отличала настоящий результат 10 от заполненного.

## Разобранный пример: далёкое значение

Ряд [8,9,10,11,1000] имеет медиану 10 и среднее 207.6. Число 1000 сильно меняет среднее, но нельзя удалять его только за необычность. Это может быть ошибка единиц, опечатка или редкий настоящий объект. **Выброс** — значение, сильно отличающееся по выбранному правилу; это сигнал к проверке, а не автоматическое доказательство ошибки.

Ограничение значений порогом, удаление строк и замена пропуском дают разные наборы данных. Если порог вычисляется из распределения, его также учат только на train. После обработки проверяют, какие группы потеряли больше объектов: фильтрация способна исказить саму задачу.

В ноутбуке SimpleImputer — преобразователь заполнения. Параметр strategy='median' выбирает медиану, add_indicator=True добавляет признаки пропусков для столбцов, где они были при обучении. Это не универсальная модель причин отсутствия, а конкретное инженерное правило.

В NumPy `np.nan` обозначает отсутствующее числовое значение. Это не измерение 0. Проверять пропуски сравнением `x == np.nan` нельзя: NaN не равен даже самому себе. Используйте `np.isnan` для числовых массивов или знакомый `isna` в pandas. В примере пропуски заменяет `SimpleImputer`, обученный только на train.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.impute import SimpleImputer
train = np.array([[8.], [10.], [12.], [np.nan]])
prepare = SimpleImputer(strategy="median", add_indicator=True)
prepare.fit(train)
new = prepare.transform([[np.nan], [100.]])
assert new.shape == (2, 2)
assert new.tolist() == [[10., 1.], [100., 0.]]
print(new)
```
