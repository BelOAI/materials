## Чему научимся

## Сначала остатки

В регрессии ответ — число. Знаковая ошибка eᵢ=прогноз−правильный ответ показывает направление ошибки. Просто усреднять знаковые ошибки нельзя: −5 и +5 дадут ноль, хотя оба прогноза неточны. **MAE** — средняя абсолютная ошибка: усреднение |eᵢ|. **MSE** — средняя квадратичная ошибка: усреднение eᵢ². **RMSE** — квадратный корень из MSE.

В [проекциях](../../../math/linear-algebra/projections/) и в бустинге используется также **остаток** r=ответ−прогноз, то есть r=−e. Например, ответ 5 и прогноз 3 дают e=−2, r=2. Модуль и квадрат от смены знака не меняются, а формула градиента меняется: перед вычислением всегда выпишите, какую разность вы обозначили буквой.

MAE и RMSE имеют единицы ответа, MSE — квадрат этих единиц. При одинаковых ошибках RMSE совпадает с MAE, при неравномерных может быть больше. Квадрат усиливает вклад больших промахов; это полезно, если именно они дороги, и опасно, если данные содержат неверные измерения.

## Разобранный пример: одинаковая сумма модулей

Для ответов [10,10] прогноз А=[8,12], Б=[10,14]. У А ошибки [−2,2], у Б [0,4]. MAE обеих моделей 2. MSE модели А равно 4, Б — 8; RMSE соответственно 2 и √8≈2.83. Одинаковая MAE скрыла разную концентрацию ошибок.

Постоянный прогноз для MSE выбирают по обучающему среднему, для MAE подходит обучающая медиана. Это baseline, который не должен использовать ответы проверочных объектов. Метрика R² сравнивает сумму квадратов ошибок с отклонениями от среднего проверочных ответов; она может быть отрицательной и не означает процент верных прогнозов. При постоянных ответах знаменатель равен нулю: требуется отдельное соглашение.

Точная формула: $R^2=1-\frac{\sum_i(\hat y_i-y_i)^2}{\sum_i(y_i-\bar y)^2}$, где $\hat y_i$ — прогноз, $y_i$ — ответ, а $\bar y$ — среднее ответов оцениваемого набора. Например, для ответов [0,2] и прогнозов [3,3] сумма квадратов ошибок равна $9+1=10$, а знаменатель $1+1=2$. Получаем $R^2=1-10/2=-4$: модель хуже постоянного среднего этого набора. Среднее проверочных ответов здесь входит только в определение метрики. Обучать на нём реальный baseline нельзя: для прогноза константу выбирают по train.

Перед сравнением убедитесь, что ответы и прогнозы относятся к одним объектам в одинаковом порядке и единицах. Ошибка в сантиметрах и та же ошибка в метрах дают разные численные MAE, хотя описывают одно расхождение.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
actual = np.array([10., 10.])
a = np.array([8., 12.])
b = np.array([10., 14.])
assert actual.shape == a.shape == b.shape
assert mean_absolute_error(actual, a) == mean_absolute_error(actual, b) == 2
assert mean_squared_error(actual, a) == 4
assert mean_squared_error(actual, b) == 8
assert r2_score([0., 2.], [3., 3.]) == -4.
print("RMSE A:", np.sqrt(mean_squared_error(actual, a)))
```
