## Слой как набор нейронов

Нейрон сначала вычисляет взвешенную сумму z=w·x+b, затем применяет **активацию** a=φ(z). Линейный слой из нескольких нейронов выдаёт вектор таких сумм. Матрица весов переводит входные признаки в новые координаты, вектор свободных членов сдвигает их.

ReLU(z)=max(0,z) сохраняет положительные значения и обнуляет отрицательные. Сигмоида переводит число в (0,1). Выбор активации зависит от роли слоя: скрытому слою нужна нелинейность, выход регрессии часто остаётся линейным. Не добавляйте сигмоиду к любому последнему слою без проверки задачи.

## Разобранный пример: два шага

Для x=(2,−1), w=(3,4), b=1 счёт равен 6−4+1=3. После ReLU выход 3. При x=(−2,1) счёт −6+4+1=−1, выход 0. Это уже нелинейное правило: разные области входа используют разные ветви.

Если два линейных слоя соединить без активации, получим A₂(A₁x+b₁)+b₂=(A₂A₁)x+(A₂b₁+b₂), то есть снова одно линейное преобразование со сдвигом. Именно промежуточная нелинейность позволяет сети описывать более сложные зависимости.

## Прямой проход

Прямой проход — вычисление от входа к выходу. Обучение ещё не произошло: случайные веса просто задают первоначальную функцию. В PyTorch nn.Sequential выполняет слои по порядку, nn.Linear задаёт линейное преобразование, nn.ReLU — активацию. Формы (n,2)→(n,3)→(n,1) сохраняют ось объектов. Все вычисления здесь на CPU.

`detach()` используется только при печати: он отделяет показанный тензор от графа, не обучая и не меняя веса.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import torch
from torch import nn
torch.set_num_threads(1)
torch.manual_seed(17)
model = nn.Sequential(nn.Linear(2, 3), nn.ReLU(), nn.Linear(3, 1))
X = torch.tensor([[2., -1.], [-2., 1.]])
output = model(X)
assert output.shape == (2, 1)
assert torch.relu(torch.tensor(-3.)).item() == 0
print(output.detach())
```
