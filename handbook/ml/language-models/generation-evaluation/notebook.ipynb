## Один красивый ответ не измерение

Генеративное решение оценивают по задаче: выполнение инструкции, фактическая точность, полнота, стиль, разнообразие и допустимость содержания. Эти критерии могут конфликтовать. Длинный ответ не обязательно полнее, а правдоподобная формулировка не гарантирует истинности.

Галлюцинацией здесь называем неподтверждённое или неверное утверждение, представленное как факт. Для справочного ответа полезно разбить текст на проверяемые утверждения и найти поддержку каждому. Отсутствие поддержки и установленная ложность — разные статусы; не смешивайте их в разметке.

## Разобранный пример: две фразы

Источник: «Библиотека работает до 18 часов». Ответ: «Библиотека открыта до 18 часов и работает по субботам». Первое утверждение подтверждается источником, второе из него не следует. На уровне ответа нельзя поставить «всё верно» только за правильный час. Если проверяем два утверждения, подтверждено одно из двух.

Для задач без единственного правильного ответа нужно заранее описать правила оценки: какие ошибки критичны, как оценивать частичное выполнение, когда допустим отказ. Разметчикам дают одинаковые инструкции, спорные случаи сохраняют. Модель-судья тоже может ошибаться и зависеть от формата; она не заменяет независимую проверку без оценки её надёжности.

## Стоимость и повторение

При цене a за входной и b за выходной токен расчёт равен a·N_in+b·N_out в согласованных единицах. К этому добавляются время, поиск, память и ручная проверка. Случайная выборка может менять качество, поэтому оценивают набор запросов и заранее выбранные повторения, а не лучший единичный ответ.

В ноутбуке используем размеченные авторские утверждения и условные цены. Внешнего API и генератора нет; код проверяет агрегирование, не устанавливает истинность автоматически.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
claims = [{"text": "до 18 часов", "status": "supported"}, {"text": "по субботам", "status": "unsupported"}]
assert len(claims) > 0
support_rate = sum([row["status"] == "supported" for row in claims]) / len(claims)
assert support_rate == 0.5
cost = 100 * 0.01 + 20 * 0.02
assert abs(cost - 1.4) < 1e-12
print("Ручная доля поддержки:", support_rate, "Условная стоимость:", cost)
```
