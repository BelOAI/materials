## Разрушение сигнала как учебная задача

Диффузионный подход постепенно добавляет шум к данным, а модель учится обращать этот процесс шаг за шагом. Один распространённый вид зашумления: $x_t=\sqrt{\bar\alpha_t}x_0+\sqrt{1-\bar\alpha_t}\,\varepsilon$, где x₀ — исходный сигнал, ε — стандартный нормальный шум, а коэффициент $0\leq\bar\alpha_t\leq1$ задаёт квадрат множителя исходного сигнала. Сам сигнал умножается на $\sqrt{\bar\alpha_t}$, шум — на $\sqrt{1-\bar\alpha_t}$. Поэтому при $\bar\alpha_t=0.25$ множитель сигнала равен 0.5, а не 0.25.

При ᾱ близком к 1 виден исходный объект, при близком к 0 преобладает шум. Индекс времени t сообщает модели уровень зашумления. Обучение часто сравнивает предсказанный шум с действительно добавленным; способы параметризации могут отличаться.

## Разобранный пример: одна координата

Пусть x₀=2, ᾱ=0.25, ε=1. Тогда xₜ=0.5·2+√0.75≈1.866. Если бы точный ε был известен и $\bar\alpha>0$, можно было бы восстановить $x_0=(x_t-\sqrt{1-\bar\alpha}\,\varepsilon)/\sqrt{\bar\alpha}$. Но при настоящей генерации исходного x₀ и добавленного ε нет: вместо них работает обученная оценка и согласованная схема обратных шагов.

Нельзя просто вычесть произвольный шум и назвать это диффузионной генерацией. Обратный процесс зависит от обученной модели, расписания и алгоритма выборки. Много обратных шагов требуют времени; ускорение может менять качество и разнообразие.

## Проверка механизма

В ноутбуке ε известен специально, чтобы проверить формулу прямого зашумления и точное алгебраическое восстановление. Это вычислительная иллюстрация, не обученный денойзер и не генератор изображений. При нулевой ᾱ обратная формула делила бы на ноль; такой граничный случай нельзя использовать для точного восстановления этим выражением.

Для прикладной оценки снова нужны разнообразие, соответствие запросу, ограничения данных и вычислительная цена. Название архитектуры не заменяет проверку результатов.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

Проверки стоят перед соответствующими действиями: диапазон коэффициента — перед корнями, ненулевой коэффициент — перед обратным делением. При нуле прямое зашумление определено, но восстановление по этой формуле невозможно.

```python
import numpy as np
x0 = np.array([1., 2., -1.])
epsilon = np.array([0.5, -0.5, 1.])
alpha_bar = 0.25
if not 0 <= alpha_bar <= 1:
    raise ValueError("Для зашумления нужен коэффициент от 0 до 1")
xt = np.sqrt(alpha_bar)*x0 + np.sqrt(1-alpha_bar)*epsilon
if alpha_bar == 0:
    raise ValueError("При нулевом коэффициенте исходный сигнал восстановить нельзя")
restored = (xt - np.sqrt(1-alpha_bar)*epsilon) / np.sqrt(alpha_bar)
assert np.allclose(restored, x0)
print("Известный шум позволяет проверить формулу:", xt, restored)
```
