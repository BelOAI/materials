## Сначала простой эксперимент

Текстовая классификация связывает документ с меткой: тема, тональность или другой заранее определённый класс. Начните с постоянного ответа и мешка слов с простой моделью. Если сложная сеть не улучшает честную проверку, её размер сам по себе не аргумент.

Pipeline из TfidfVectorizer и LogisticRegression учит словарь, idf и коэффициенты только по train. Вероятности и порог рассматриваются отдельно. Случайное перемешивание почти одинаковых отзывов может дать высокую оценку за узнавание копий.

## Разобранный пример: знакомая фраза

В train есть «очень хороший урок», в test — «Очень хороший урок!». Простая нормализация показывает почти дубликат. Если цель — новые авторы или новые темы, такие записи нужно группировать по происхождению до разбиения. Совпадение текста не единственный вид зависимости: разные фразы одного шаблона также могут выдавать источник.

Эмбеддинг документа — другой способ получить вектор. Его можно усреднить из представлений токенов или обучить специальной моделью. Но полезность проверяется той же задачей; вектор не становится «семантическим» только от названия. Если представление обучалось с ответами, обучение входит внутрь допустимого train.

## Прозрачные ограничения

В ноутбуке восемь коротких авторских фраз и два запроса. Мы проверяем путь преобразований и количество ответов, не объявляя такую выборку достаточной оценкой качества русского языка. Слова, явно связанные с меткой, делают пример удобным для механики, но не заменяют разнообразный независимый набор.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
texts = ["хороший урок", "понятный урок", "полезный пример", "хороший пример", "плохой урок", "непонятный урок", "скучный пример", "плохой пример"]
y = [1, 1, 1, 1, 0, 0, 0, 0]
model = Pipeline([("text", TfidfVectorizer()), ("classify", LogisticRegression(random_state=17))])
model.fit(texts, y)
p = model.predict_proba(["хороший урок", "плохой урок"])
assert p.shape == (2, 2)
assert p[0, 1] > p[1, 1]
print(p)
```
