## Чему научимся

## От строки к единицам

Токен — выбранная единица обработки текста: слово, часть слова или символ. **Токенизация** задаёт правило разбиения. Простой split по пробелам полезен для первого примера, но не решает автоматически пунктуацию, сложные языки и составные слова.

Нормализация может включать нижний регистр и удаление крайних пробелов. Каждое правило имеет последствия: «Мир» как название и «мир» как слово могут различаться по смыслу. Знаки отрицания нельзя удалять только потому, что они короткие.

## Разобранный пример: свой словарь

Train содержит «кот спит» и «кот играет». После нижнего регистра словарь из трёх слов: играет, кот, спит. Назначим им индексы 1,2,3, а 0 оставим для неизвестного токена. Новое сообщение «кот бежит» превращается в [2,0]. Неизвестное слово не исчезло молча, но его точное содержание потеряно.

Словарь не нужно расширять по тестовым сообщениям при обычном независимом протоколе. Иначе подготовка уже знает состав проверки. Подсловные токенизаторы уменьшают долю полностью неизвестных слов, но имеют собственные правила, которые нельзя бездумно менять у уже обученной модели.

## Что сохранять

Нужны само правило обработки, словарь и соответствие индексов. Одинаковый набор слов с другой нумерацией несовместим с весами модели. Специальные индексы для заполнения и неизвестного значения также не должны случайно совпадать, если выполняют разные роли.

В ноутбуке используем только стандартную библиотеку и маленькие авторские фразы. Это не промышленный языковой токенизатор, а прозрачный пример пути от текста к числам.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
train = ["кот спит", "кот играет"]
unique_words = set()
for text in train:
    for word in text.lower().split():
        unique_words.add(word)
words = sorted(unique_words)
vocabulary = {word: index + 1 for index, word in enumerate(words)}
def encode(text):
    return [vocabulary.get(word, 0) for word in text.lower().split()]
assert encode("кот бежит") == [vocabulary["кот"], 0]
assert "бежит" not in vocabulary
print(vocabulary, encode("кот бежит"))
```
