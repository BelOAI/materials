## Генерация через скрытое описание

Скрытая, или латентная, переменная z — компактное описание, из которого декодер строит объект. В обычном автоэнкодере encoder сжимает x в z, decoder восстанавливает x. Хорошее восстановление train ещё не гарантирует, что произвольные z дадут осмысленные новые объекты.

**VAE**, вариационный автоэнкодер, задаёт распределение скрытого описания, часто с параметрами среднего μ и стандартного отклонения σ. Выбор z=μ+σε, где ε — стандартный нормальный шум, отделяет случайность от обучаемых параметров. Цель сочетает качество восстановления и штраф, приближающий распределение скрытых описаний к выбранному простому распределению.

## Разобранный пример: скрытый шум

При μ=2,σ=0.5 и ε=−1 получается z=1.5. При ε=1 — 2.5. Один исходный объект не обязан иметь одно единственное случайное скрытое значение. Штраф KL сравнивает распределения, а не расстояние между отдельными z. Баланс восстановления и штрафа влияет на сохранённую информацию.

Например, условная цель «ошибка восстановления + λ·KL» у варианта А равна $1+0.5\cdot4=3$, у Б — $2+0.5\cdot1=2.5$. При λ=0.5 предпочтительнее Б, хотя восстанавливает он хуже. Это учебные значения двух частей цели, а не универсальная формула ошибки изображения. В VAE распределение скрытого описания зависит от входа x: encoder вычисляет его μ и σ, а KL сравнивает это распределение с заранее выбранным, обычно стандартным нормальным. Для непрерывных распределений KL определяют интегралом вместо суммы; его смысл дополнительного штрафа сохраняется.

## Другой подход: GAN

GAN содержит генератор G, превращающий шум в объект, и дискриминатор D, различающий реальные и сгенерированные примеры. Генератор старается делать объекты, которые D принимает за реальные; D учится не путать их с исходными данными. Это состязание двух обучаемых частей, а не один фиксированный критерий качества.

Если генератор выдаёт один красивый объект независимо от шума, возникает потеря разнообразия, или mode collapse. Низкая ошибка дискриминатора не равна хорошей генерации. Для оценки нужны разнообразие, соответствие условию и отсутствие копирования train.

В ноутбуке показываем только выборку скрытой переменной и контрпример схлопнутого генератора. Ни VAE, ни GAN здесь не обучены; небольшой расчёт не выдаётся за полноценную генерацию изображений.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
rng = np.random.default_rng(17)
mu, sigma = 2., 0.5
epsilon = rng.normal(size=6)
z = mu + sigma * epsilon
assert z.shape == (6,)
assert sigma > 0
noise = np.arange(6.)
collapsed_outputs = np.full_like(noise, 7.)
assert len(np.unique(collapsed_outputs)) == 1
print("Скрытые значения:", z, "Контрпример разнообразия:", collapsed_outputs)
```
