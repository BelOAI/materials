## Мешок потерял порядок

Фразы «день сменяет ночь» и «ночь сменяет день» состоят из одних и тех же слов с теми же частотами, но сообщают о противоположных сменах. Мешок слов даст им одинаковые признаки: он потерял порядок. Последовательная модель обрабатывает токены по очереди и обновляет состояние hₜ=f(hₜ₋₁,xₜ). Здесь t — номер токена, начальное h₀ задают отдельно, xₜ — вектор текущего токена, hₜ — состояние после его обработки. Один и тот же токен может привести к разным состояниям в зависимости от того, что встретилось раньше.

Простая рекуррентная сеть использует hₜ=tanh(Ahₜ₋₁+Bxₜ+b). Tanh ограничивает координаты между −1 и 1. Параметры повторно используются на всех позициях. В отличие от отдельного свободного веса на каждый момент, это позволяет применять правило к разной длине.

Индекс токена в словаре — только его номер, а не измерение смысла. Для сети этот номер связывают с вектором, например строкой таблицы эмбеддингов. В ручном расчёте ниже вместо таких векторов возьмём числа 1 и 2, чтобы увидеть влияние порядка без длинной матричной арифметики.

## Разобранный пример: память со затуханием

Для прозрачности возьмём не обученную сеть, а рекуррентное правило hₜ=0.5hₜ₋₁+xₜ, h₀=0. Последовательность [1,2] даёт h₁=1,h₂=2.5. Переставленная [2,1] даёт 2, затем 2. Слова те же по набору, состояние различается из-за порядка.

Вклад давнего входа умножается на степень 0.5. Через много шагов он мал; при других весах значения и градиенты могут, наоборот, расти. Долгие зависимости затрудняют обучение простой рекурсии. Это одна из причин использовать механизмы управления памятью и внимание.

## Что считать выходом

Для классификации можно использовать последнее состояние; для прогноза следующего токена — выход на каждой позиции. Заполнение коротких последовательностей до общей длины не должно создавать ложные значимые токены: нужна маска действительных позиций или учёт длины.

Численный пример показывает принцип рекуррентного обновления состояния; он не обучает языковую модель и не запускает LLM. Следующий урок позволит каждой позиции обращаться к другим через внимание.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
def state(sequence):
    h = 0.
    trace = []
    for value in sequence:
        h = 0.5*h + value
        trace.append(h)
    return trace
assert state([1., 2.]) == [1., 2.5]
assert state([2., 1.]) == [2., 2.]
assert state([]) == []
print(state([1., 0., 0., 0.]))
```
