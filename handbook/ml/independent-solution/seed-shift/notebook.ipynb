## Повторение одного протокола

Seed меняет случайный старт, перемешивание и иногда разбиение. Чтобы понять чувствительность решения, заранее задают несколько seed и сохраняют результаты всех. Сравнивать модели полезно попарно на одинаковых частях: тогда разность меньше смешивается с разной трудностью проверки.

Среднее и разброс трёх запусков — описание этих запусков, а не автоматический доверительный интервал для всех возможных условий. Если seed выбирался по тесту, повторения уже стали способом подбора.

## Разобранный пример: устойчивое улучшение

Модель А получила [0.70,0.72,0.68], Б — [0.71,0.73,0.69] на трёх согласованных запусках. Во всех парах разность 0.01, средние 0.70 и 0.71. Это убедительнее одного удачного Б=0.80 на другом, более лёгком разбиении, но объём данных и ошибки протокола всё ещё важны.

## Сдвиг данных

Новые объекты могут отличаться по признакам, долям классов или связи признаков с ответом. Это **сдвиг распределения**. Повторение случайных seed внутри старого набора не моделирует автоматически новый город, новый год или другой прибор.

Проверяйте покрытие групп, диапазоны, пропуски и фактическое время появления сведений. Временная проверка лучше отвечает на вопрос о будущем, чем множество случайных разделений прошлого. Не нужно автоматически переучивать подготовку на тесте для «исправления» сдвига: это меняет протокол применения.

В ноутбуке отдельно считаем парные изменения и изменение среднего нового признака. Последнее — диагностический сигнал, а не доказательство, какой именно тип сдвига произошёл и насколько упало качество.

В [Python](../../../python/ml-tools/reproducibility/) мы сохраняли данные, разбиение и окружение, чтобы повторить один запуск. Здесь вопрос шире: повторяется ли вывод о преимуществе метода при нескольких заранее выбранных стартах и на новых условиях данных? Воспроизводимость одного опыта и устойчивость вывода — разные проверки.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
import numpy as np
a = np.array([0.70, 0.72, 0.68])
b = np.array([0.71, 0.73, 0.69])
assert np.isclose(a.mean(), 0.70)
assert np.allclose(b-a, 0.01)
old = np.array([1., 2., 3.])
new = np.array([10., 11., 12.])
assert new.mean() > old.mean()
print("Парное улучшение:", (b-a).mean(), "Изменение среднего признака:", new.mean()-old.mean())
```
