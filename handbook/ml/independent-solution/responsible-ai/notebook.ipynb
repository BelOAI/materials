## Найдено не значит разрешено

Публично открытая страница может содержать личные сведения или материал с ограниченными правами. До использования выясните источник, автора, лицензию, условия задачи и допустимые действия. Правила на данные, программный код и веса модели могут различаться. Этот урок учит проверять условия, а не подменяет их конкретной юридической консультацией.

Минимизация данных означает собирать и хранить только нужные для задачи сведения. Удаление имени не всегда делает запись анонимной: сочетание школы, возраста и редкого события может позволить узнать человека. Сохранённый журнал также может случайно раскрыть исходный текст или идентификатор.

## Разобранный пример: два набора

Для собственных вымышленных фраз известны автор и способ создания; можно описать их учебный статус и лицензию. Для найденной таблицы с телефонами наличие ссылки недостаточно: происхождение, права и необходимость личных полей не установлены. Корректное действие — приостановить использование такого источника, а не скопировать его в ноутбук и забыть.

Для внешней модели отдельно проверяют происхождение весов, условия использования и доступность исходных данных, насколько это известно. Нельзя обещать полноту проверки, если часть происхождения не раскрыта.

## Запись решения

Для каждого ресурса полезна краткая карточка: откуда получен, когда, кем создан, какие условия проверены, какие ограничения остались. В этой практике используем авторские синтетические данные с явно описанным происхождением.

В ноутбуке сравниваем два вымышленных описания ресурсов по явно заданным полям. Код не устанавливает юридическую законность автоматически: он лишь обнаруживает недостающие сведения для дальнейшего решения.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
resources = [
    {"name": "авторские фразы", "source": "созданы для урока", "terms": "CC BY 4.0", "personal": False},
    {"name": "неизвестная таблица", "source": "", "terms": "", "personal": True},
]
def missing_information(row):
    return [field for field in ["source", "terms"] if not row[field]]
assert missing_information(resources[0]) == []
assert missing_information(resources[1]) == ["source", "terms"]
assert resources[1]["personal"]
print("Поля для дополнительной проверки:", missing_information(resources[1]))
```
