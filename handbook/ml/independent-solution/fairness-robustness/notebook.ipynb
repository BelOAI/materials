## Общая метрика не описывает всех

Модель может одинаково часто угадывать в целом, но пропускать большую долю положительных объектов одной группы. Поэтому сравнивают конкретные виды ошибок и размеры групп. Разные определения справедливости могут требовать несовместимых равенств при разных исходных долях; один общий флаг «справедливо» не заменяет постановку цели.

Смещение может возникнуть при сборе, разметке, определении ответа и выборе порога. Удаление названия группы не устраняет её косвенные признаки. При этом анализ групп сам требует аккуратного и допустимого использования сведений.

## Разобранный пример: разная полнота

В группе А из десяти положительных найдены восемь, в Б из пяти — два. Recall равен 0.8 и 0.4. Это сигнал для исследования причин: представительство train, качество меток, условия измерения. По таким маленьким числам нельзя уверенно оценить точную величину устойчивого разрыва.

**Робастность**, устойчивость, проверяет поведение при допустимых изменениях входа: небольшом шуме, пропуске, смене освещения. Изменение должно сохранять правильный ответ. Поворот цифры или удаление отрицания не всегда допустимы как «маленькое безвредное изменение».

## Когда нужен человек

Определите ситуации низкой уверенности, отсутствия сведений и высокой цены ошибки. Передача человеку полезна только с понятными данными и ответственностью, а не с формальным требованием подтвердить непрозрачный ответ. Ошибки генерации также требуют проверки источников и возможности отказа.

В ноутбуке считаем recall двух вымышленных групп и меняем порог числового правила рядом с границей. Это диагностика конкретного поведения, не универсальный сертификат справедливости.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
groups = {"A": {"tp": 8, "fn": 2}, "B": {"tp": 2, "fn": 3}}
recall = {key: row["tp"] / (row["tp"] + row["fn"]) for key, row in groups.items()}
assert recall == {"A": 0.8, "B": 0.4}
def decision(x):
    return int(x >= 1.)
assert decision(0.99) != decision(1.01)
print(recall, "Точки около границы требуют анализа устойчивости")
```
