## Где модель ошибается

Общая оценка усредняет разные режимы. Разделите ошибки по заранее осмысленным срезам: новые и знакомые пользователи, длина текста, диапазон освещения, редкие классы. Для каждого среза сообщайте размер. Одна ошибка в группе из двух объектов не имеет той же определённости, что половина ошибок среди тысячи.

Не создавайте сотни срезов, чтобы выбрать единственный эффектный. Это ещё один вид множественного поиска. Проверяемая гипотеза должна объяснять, почему изменение решения поможет конкретному виду ошибок.

## Разобранный пример: общий результат скрыл проблему

В группе А 90 объектов и 9 ошибок, в Б 10 объектов и 5 ошибок. Общая доля ошибок 14/100=0.14. В А она 0.1, в Б — 0.5. Уменьшение ошибок А на одну не решает провал Б, хотя общая метрика улучшится.

**Ablation**, исключение компонента, сравнивает полное решение с вариантом без выбранной детали. Если удаление признака улучшило validation, это повод проверить утечку или шум, а не объявлять признак вредным навсегда. Остальные условия, разбиение и бюджет должны быть сопоставимы.

Кривая обучения показывает зависимость качества от размера train. Подвыборки создают внутри разработки; итоговый тест не используют для многократной диагностики. Разницу метрик сопровождают её масштабом относительно случайной изменчивости.

В ноутбуке анализируем заранее заданные ошибки, а не получаем их обучением. Данные придуманы, чтобы арифметика срезов была прозрачной.

## Разобранный пример: проверяем вычисление

Все числа в примере придуманы для этого урока. Выполните код, затем измените вход и объясните новый результат.

```python
rows = [{"group": "A", "errors": 9, "count": 90}, {"group": "B", "errors": 5, "count": 10}]
for row in rows:
    if row["count"] <= 0:
        raise ValueError("Для доли ошибок в срезе нужны наблюдения")
    assert 0 <= row["errors"] <= row["count"]
    print(row["group"], row["errors"] / row["count"], row["count"])
overall = sum([r["errors"] for r in rows]) / sum([r["count"] for r in rows])
assert overall == 0.14
```
