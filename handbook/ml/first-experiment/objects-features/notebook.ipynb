## Задача и происхождение данных

Представим мастерскую бумажных планеров. Для каждого планера записаны размах крыла и масса. Нужно предсказать класс полёта: `короткий` или `дальний`. Это **классификация** — выбор одной метки из конечного набора. Предсказание численного расстояния полёта было бы задачей **регрессии**.

В нашем [наборе данных](../../../data/gliders.csv) 24 придуманных объекта. Он специально устроен просто и не описывает физику полёта. [Описание генерации и полей](../../../data/README.md) объясняет все числа. В реальном исследовании нужны измерения, условия запуска и определение границы классов.

| id | span_cm | mass_g | flight | split |
| --- | --- | --- | --- | --- |
| g01 | 12 | 8 | короткий | train |
| g02 | 22 | 10 | дальний | train |
| g03 | 14 | 28 | короткий | train |
| g04 | 24 | 30 | дальний | train |

**Объект** — одна строка. **Признаки** — величины, которые доступны перед предсказанием: `span_cm` — размах в сантиметрах, `mass_g` — масса в граммах. В таблице показаны первые четыре строки из 24. **Целевая переменная** — правильный ответ `flight`. Идентификатор `id` нужен для различения строк, а `split` — для организации эксперимента; модель их не получает.

В Python словом «объект» называют также значение в памяти, например список или словарь. Здесь объект данных — наблюдаемый планер; программа может представить его словарём. Его упорядоченный список числовых признаков будем называть вектором. Геометрию векторов разберём перед KNN; сейчас достаточно сохранять порядок полей.

## Что доступно перед предсказанием

Признак должен быть известен в момент применения модели. Размах и масса можно измерить до полёта. Итоговую дальность полёта нельзя использовать для предсказания класса этого же полёта: она уже раскрывает ответ. Сведения, появляющиеся после события, делают проверку обманчиво лёгкой.

Идентификатор нужен, чтобы сопоставить строки, но его числовой порядок может случайно совпасть с классом. Такой порядок не обязательно сохранится у новых объектов. Поэтому начнём только с измерений.

## Разобранный пример: запись одного объекта

```python
row = {"id": "g01", "span_cm": 12, "mass_g": 8, "flight": "короткий"}
features = [row["span_cm"], row["mass_g"]]
answer = row["flight"]
assert features == [12, 8]
assert answer == "короткий"
```

Порядок двух признаков становится частью договорённости: для всех объектов сначала размах, затем масса. Правильный ответ хранится отдельно. Предсказание числовой дальности потребовало бы другой цели и другого способа измерить ошибку.

## Учебные данные и реальные выводы

Для реальных измерений фразу «дальний полёт» нужно превратить в правило, например «пролетел не меньше пяти метров». Иначе два человека могут подписать одинаковую дальность разными метками. Правило фиксируют до сбора ответов; этот порог приведён только для примера и не задаёт метки в нашем синтетическом наборе.

## Разобранный пример: проверяем доступность

Нужно до запуска планера предсказать дальность. Размах и масса измерены до запуска и подходят как кандидаты. Время полёта и отметка судьи появляются после запуска и для этого прогноза недоступны. Температура может подходить, если измерена в нужный момент и будет доступна для новых объектов.

Пригодность столбца зависит от момента прогноза. Время полёта можно использовать для описания уже завершённого запуска, но нельзя — для прогноза до старта. Поэтому сначала определите, что и в какой момент предсказываете, затем выбирайте признаки.

## Типичные ошибки

Не включайте ответ и сведения после события в признаки раннего прогноза. Совпадение идентификатора с меткой на маленькой таблице не доказывает полезного закона.
