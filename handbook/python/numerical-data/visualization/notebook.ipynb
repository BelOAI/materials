## Таблица для примеров

Вернёмся к четырём вымышленным планерам из работы с таблицами. Теперь вопрос другой: как увидеть распределение размаха и связь размаха с массой? Восстановим таблицу, чтобы ноутбук запускался отдельно; пропущенная масса третьего планера повлияет на число точек графика.

```python
import pandas as pd
frame = pd.DataFrame({
    "id": ["a", "b", "c", "d"],
    "span_cm": [12., 22., 14., 24.],
    "mass_g": [8., 10., None, 12.],
    "flight": ["короткий", "дальний", "короткий", "дальний"],
})
```

Для графиков подключим модуль `pyplot` библиотеки matplotlib под коротким именем `plt`.

```python
import matplotlib.pyplot as plt
```

## График отвечает на вопрос

Гистограмма показывает распределение одного числового признака по интервалам, диаграмма рассеяния — связь двух признаков, столбчатая диаграмма — сравнение категорий. Видимая связь на графике сама по себе не доказывает причинную связь: обе величины могут меняться из-за третьей. Одинаковые средние тоже не гарантируют похожего разброса: у наборов `[10, 10]` и `[1, 19]` одно среднее 10, но очень разные значения.

`plt.subplots(2, 1, figsize=(5, 6))` создаёт рисунок `fig` и две области графиков `axes` друг под другом; `figsize` задаёт размер в дюймах. Метод `hist` строит гистограмму, `bins` задаёт границы её интервалов, `scatter` — диаграмму рассеяния. Методы `set_xlabel` и `set_ylabel` подписывают оси, `set_xticks` и `set_yticks` задают положения делений. Для числа объектов выбираем целые деления. `tight_layout` подбирает отступы, `show` показывает рисунок, `close` освобождает его после показа.

```python
fig, axes = plt.subplots(2, 1, figsize=(5, 6))
axes[0].hist(frame["span_cm"], bins=[10, 15, 20, 25])
axes[0].set_xlabel("Размах, см")
axes[0].set_ylabel("Число объектов")
axes[0].set_xticks([10, 15, 20, 25])
axes[0].set_yticks([0, 1, 2])
axes[1].scatter(frame["span_cm"], frame["mass_g"])
axes[1].set_xlabel("Размах, см")
axes[1].set_ylabel("Масса, г")
fig.tight_layout()
plt.show()
plt.close(fig)
```

Интервалы гистограммы здесь — `[10, 15)`, `[15, 20)`, `[20, 25]`: левая граница входит, правая не входит, кроме последнего интервала. Поэтому 15 попало бы во второй столбец. Ожидаем столбцы высотой 2, 0, 2. На диаграмме рассеяния будет три точки: строка с пропущенной массой не рисуется. Подписи с единицами позволяют понять график без чтения кода. Четырёх вымышленных объектов недостаточно для выводов о реальных планерах.

## Разобранный пример: сравнение категорий

Для числа объектов каждого класса подходит столбчатая диаграмма. Сначала считаем строки, затем передаём подписи и значения в `bar`. В отличие от гистограммы, столбцы здесь соответствуют категориям, а не числовым интервалам.

```python
counts = frame.groupby("flight").size()
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(counts.index.tolist(), counts.tolist())
ax.set_xlabel("Класс полёта")
ax.set_ylabel("Число объектов")
ax.set_yticks([0, 1, 2])
fig.tight_layout()
plt.show()
plt.close(fig)
```

Два столбца имеют высоту 2. Пропуск класса или удаление строк перед рисунком изменит смысл сравнения. Всегда подписывайте единицы и проверяйте, какие строки попали на график. Для вывода о связи размаха и массы четырёх придуманных объектов недостаточно.
