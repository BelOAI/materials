## Массив, форма и тип

NumPy — библиотека числовых вычислений. Модуль — набор готовых инструментов. Запись `import numpy as np` подключает NumPy под коротким именем `np`; вызовы начинаются с этого имени. Библиотеку нужно предварительно установить, как описано во введении. `np.array` создаёт массив из списка или вложенных списков. Рассматриваем числовые массивы: все их элементы имеют общий тип.

Свойство (атрибут) — значение, читаемое через точку без круглых скобок; метод вызывается со скобками. Свойство `shape` — кортеж длин осей, `ndim` — число осей, `dtype` — тип элемента, `nbytes` — число байт элементов без служебных данных. У таблицы ось 0 перечисляет строки, ось 1 — столбцы. Форма `(3, 2)` означает три строки и два столбца; `(3,)` — одну ось длины 3. Запятая в `(3,)` нужна для кортежа с одним элементом.

В [матричной записи](../../../math/linear-algebra/matrices-products/) X имеет n строк-объектов и d столбцов-признаков; в NumPy это `X.shape == (n, d)`. В формулах индексы обычно идут с 1, в Python — с 0: элемент $X_{ij}$ соответствует `X[i - 1, j - 1]`. Вектор длины d здесь имеет форму `(d,)`, а столбец из математики — `(d, 1)`. Это разные формы: для записи вектора столбцом используйте `reshape(d, 1)`, когда этот метод будет введён в следующем уроке.

Тип `np.float64` занимает 8 байт на число и хранит дробные значения приближённо. Целые типы вроде `int8` имеют ограниченный диапазон: слишком большой результат может переполниться. Это отличается от обычных целых Python. Выбирайте тип по диапазону, а не только ради экономии памяти.

```python
import numpy as np

X = np.array([[12, 8], [22, 10], [14, 9]], dtype=np.float64)
assert X.shape == (3, 2)
assert X.ndim == 2
assert X.nbytes == 3 * 2 * 8
```

Строки — три вымышленных объекта, столбцы — размах в сантиметрах и масса в граммах. Например, `X[0, 1]` равно 8: строка 0, столбец 1.

## Среднее и сумма по осям

Методы `mean` и `sum` вычисляют среднее и сумму. Параметр `axis` указывает, какую ось убрать вычислением: `axis=0` объединяет строки и оставляет результат для каждого столбца, `axis=1` объединяет столбцы и оставляет результат для каждой строки. Без `axis` операция использует все элементы.

```python
print(X.mean(axis=0))  # [16.  9.]
print(X.sum(axis=1))   # [20. 32. 23.]
```

Первое среднее равно `(12 + 22 + 14) / 3 = 16`. Вторая строка вывода содержит суммы размаха и массы; это лишь демонстрация оси, у такой суммы разных единиц нет полезного физического смысла. Для пустого набора среднее не определено; проверяйте число строк заранее.

## Разобранный пример: один показатель на объект

Пусть строки содержат три результата одного ученика. Среднее по строке должно дать по одному числу на ученика: ось столбцов исчезнет, форма результата станет `(2,)`. Для среднего по каждому заданию выбираем ось строк.

```python
scores = np.array([[2., 4., 6.], [3., 6., 9.]])
assert scores.shape == (2, 3)
print(scores.mean(axis=1))  # [4. 6.]
print(scores.mean(axis=0))  # [2.5 5.  7.5]
```

Средние по разным осям отвечают на разные вопросы. Совпадение количества строк и столбцов может скрыть ошибку формы, поэтому проверяйте пример с прямоугольной таблицей.
