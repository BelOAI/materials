## Размер входа и порядок роста

Пусть $n$ — число объектов. Суммирование списка делает примерно $n$ сложений. Для пар разных объектов без учёта порядка первый объект образует $n-1$ пар со следующими, второй — $n-2$, последний — ни одной. Всего $(n-1)+(n-2)+\cdots+1=n(n-1)/2$ проверок. При увеличении входа в 10 раз первая работа растёт примерно в 10 раз, вторая — примерно в 100.

Запись $O(n)$ означает верхнюю оценку роста с точностью до постоянного множителя при достаточно большом $n$. Это не секунды. $O(1)$ — постоянное число действий, $O(n^2)$ — квадратичный рост, $O(n\log n)$ обычно встречается у сортировки сравнением. Основание логарифма не меняет такой класс роста. Здесь предполагаем, что сравнение или сложение небольших чисел занимает постоянное время.

Время зависит также от кода программы, компьютера и формата данных. Замер на одном маленьком входе не доказывает оценку роста для больших входов; такую оценку называют асимптотикой. Обозначение $O$ само по себе также не обещает точного времени: линейный алгоритм может иметь большую постоянную стоимость.

## Поиск и сортировка

Линейный поиск в худшем случае смотрит все $n$ элементов. `sorted` создаёт новый список; метод `list.sort()` меняет существующий. Сортировка Python имеет худшую временную оценку $O(n\log n)$.

```python
def find_index(values, target):
    for index, value in enumerate(values):
        if value == target:
            return index
    return None

assert find_index([4, 8, 4], 8) == 1
assert find_index([], 8) is None
ordered = sorted([8, 4, 4])
print(ordered)  # [4, 4, 8]
```

В отсортированном списке можно искать двоичным поиском: сравнить с серединой и оставить половину диапазона. После $k$ шагов остаётся примерно $n/2^k$ элементов, поэтому нужно $O(\log n)$ сравнений. Но предварительная сортировка стоит времени и меняет порядок. Для одного поиска она может быть лишней.

Модуль — набор готовых инструментов Python; запись `from bisect import bisect_left` подключит нужную функцию. `bisect_left` из стандартного модуля `bisect` ищет позицию вставки значения слева от равных ему элементов, сохраняя сортировку. Он не проверяет наличие значения: результат может быть равен длине списка, то есть указывать сразу за последним элементом. Поэтому сначала проверяем границу, затем равенство.

```python
from bisect import bisect_left

index = bisect_left(ordered, 8)
assert index < len(ordered) and ordered[index] == 8
```

## Хеширование и подсчёт

Словарь и множество используют хеширование: по значению ключа вычисляют число — хеш, которое помогает выбрать место поиска. Если разные ключи получили одинаковый хеш, их дополнительно различают сравнением. Для обычных ключей проверка наличия занимает в среднем $O(1)$, но это не гарантия худшего случая. Частоты можно собрать за ожидаемое $O(n)$ время и $O(u)$ дополнительной памяти, где $u$ — число разных ключей.

## Разобранный пример: существует ли пара с заданной суммой

```python
def has_pair(values, target):
    seen = set()
    for value in values:
        if target - value in seen:
            return True
        seen.add(value)
    return False

assert has_pair([12, 22, 14], 34)
assert not has_pair([17], 34)
assert has_pair([17, 17], 34)
```

Перед шагом `seen` содержит только предыдущие значения. Поэтому один объект не используется дважды; второе значение 17 может образовать пару с первым. Ожидаемое время $O(n)$, дополнительная память $O(n)$. Перебор пар использует $O(n^2)$ времени, но может обходиться постоянной дополнительной памятью, если нужен лишь ответ «да/нет».

## Большие таблицы и память

Байт — единица объёма памяти. В следующих оценках GB означает миллиард байт; служебные данные самой структуры считаются отдельно. Плотная таблица хранит каждую ячейку, включая нулевые.

Числовая таблица из миллиона строк и 20 столбцов с 8 байтами на число занимает около 160 миллионов байт без служебных данных. Списки Python обычно требуют заметно больше. Копирование таблицы удваивает занятую ею память, а таблица всех попарных расстояний содержит уже $n^2$ чисел.

Если требуется только сумма, читайте строки последовательно и храните накопитель. Если требуются все пары расстояний, подумайте об обработке блоками. Не выделяйте огромный массив ради учебного замера: сначала оцените размер формулой.
