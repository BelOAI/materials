## Значения и имена

Числа, строки и логические значения имеют разные типы. Строка записывается в кавычках; `True` и `False` означают истину и ложь. В Python десятичная часть отделяется точкой. Знак `=` связывает имя слева со значением справа; такое имя называют переменной. Комментарий начинается с `#` и не выполняется. Сравнение `<` проверяет «меньше» и возвращает логическое значение.

````python
span = 22             # целое число: размах в сантиметрах
mass = 10.5           # число с дробной частью: масса в граммах
flight = "дальний"    # строка
is_light = mass < 15  # логическое значение
print(span, flight, is_light)
````

Вывод: `22 дальний True`. Имя можно использовать справа в новом выражении: сначала вычисляется правая часть с прежним значением `span`, затем имя связывается с результатом.

````python
span = span + 2
print(span)  # 24
````

`"22"` — строка, а `22` — число. `int("22")` превращает строку в целое число. Неверное преобразование, например `int("крыло")`, вызывает ошибку. Проверяйте смысл входных данных, а не только возможность преобразовать их.

## Арифметика и запись результата

Знаки `+`, `-`, `*` обозначают сложение, вычитание и умножение.

`/` делит, `//` даёт целую часть с округлением вниз, `%` — остаток, `**` возводит в степень. Для отрицательных чисел округление вниз отличается от отбрасывания дробной части: `-7 // 2` равно −4, `-7 % 2` равно 1. На ноль делить нельзя. Умножение и деление выполняются раньше сложения; скобки делают порядок явным.

````python
print(7 + 2, 7 - 2, 7 * 2)  # 9 5 14
print(7 / 2, 7 // 2, 7 % 2) # 3.5 3 1
print(2 ** 3)               # 8
print((2 + 3) * 4)          # 20
````

F-строка начинается с `f` перед кавычкой. В ней выражения в фигурных скобках заменяются значениями, а `:.2f` задаёт два знака после точки при выводе, не меняя само число:

````python
value = 3 / 4
print(f"Доля верных: {value:.2f}")  # Доля верных: 0.75
````

Дробные числа хранятся приближённо; способы сравнения с допуском изучим в NumPy.

## Ввод и вывод

`print` выводит значения через пробел и завершает строку. Функция `input` показывает приглашение в скобках и ожидает строку с клавиатуры. `float` преобразует строку с числом в дробное число; десятичный разделитель — точка. При вводе `12.5` вывод будет `Размах: 12.5`. `input` всегда возвращает строку: сложение `"12" + "3"` даёт `"123"`, а не 15. Сначала преобразуйте данные в число. `int` читает целое, `float` — дробное с точкой. Запись `int("12.5")` неверна; `float("12.5")` допустима.

Найдите указанную строку, проверьте значения и повторите запуск. Код после необработанной ошибки не выполняется. Подробно разбирать ошибки и проверять функции будем в отдельной главе.

## Списки и циклы

Список хранит последовательность значений, записанных через запятую в квадратных скобках. Индекс — номер элемента: индексы начинаются с нуля. Обращение `spans[0]` берёт первый элемент, `len(spans)` возвращает число элементов. Пустой список записывается `[]`.

````python
spans = [12, 22, 14, 24]
print(spans[0], len(spans))  # 12 4
````

Цикл `for` выполняет действие для каждого элемента. Отступы обязательны: строки с одинаковым отступом составляют блок. В примерах используем четыре пробела.

````python
total = 0
for value in spans:
    total = total + value
mean = total / len(spans)
print(mean)  # 18.0
````

Для пустого списка такое среднее не определено: деление на ноль вызовет ошибку.

## Условия и функции

`if` выбирает действие по логическому условию. Сравнения `>=` и `<=` означают «не меньше» и «не больше», `>` — «больше», `==` — «равно», `!=` — «не равно». Знак `=` присваивает значение, а `==` сравнивает их. Если условие ложно, выполняется блок `else`.

````python
for value in spans:
    if value >= 20:
        print("большой размах")
    else:
        print("малый размах")
````

Функция объединяет действия под именем. Ключевое слово `def` начинает определение; в скобках записываются параметры — имена входных значений. При вызове, например `average(spans)`, вместо параметра `values` используется переданный список. `return` возвращает результат и завершает вызов. `raise ValueError(...)` останавливает вычисление с сообщением о неверном входе. Без `return` функция возвращает специальное значение `None`, означающее отсутствие результата.

````python
def average(values):
    if len(values) == 0:
        raise ValueError("Нужен непустой список")
    total = 0
    for value in values:
        total = total + value
    return total / len(values)

print(average(spans))  # 18.0
````

## Разобранный пример: доля верных ответов

Класс — название группы, например «короткий полёт». Модель здесь — программа, выдающая такой ответ для объекта. Пусть известные классы четырёх объектов — `короткий`, `дальний`, `короткий`, `дальний`. Модель дала ответы `короткий`, `дальний`, `дальний`, `дальний`. Верны три ответа из четырёх: доля равна 0.75. В условии функции `or` означает «или»: достаточно, чтобы хотя бы одна проверка оказалась истинной. `range(4)` задаёт числа 0, 1, 2, 3; поэтому `range(len(expected))` перебирает все индексы списка.

````python
def accuracy(expected, predicted):
    if len(expected) == 0 or len(expected) != len(predicted):
        raise ValueError("Нужны непустые списки одинаковой длины")
    correct = 0
    for index in range(len(expected)):
        if expected[index] == predicted[index]:
            correct = correct + 1
    return correct / len(expected)

actual = ["короткий", "дальний", "короткий", "дальний"]
answers = ["короткий", "дальний", "дальний", "дальний"]
print(accuracy(actual, answers))  # 0.75
````

Выражение `len(expected)` вычисляется до вызова `range`. Мы проверили длины заранее, поэтому оба списка имеют элемент на каждом индексе.
