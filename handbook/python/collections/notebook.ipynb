## Последовательности: строки, списки и кортежи

Пустой кортеж — `()`, кортеж с одним элементом — `(3,)`: запятая отличает его от числа в скобках.

Строка `str` хранит текст, список `list` — изменяемую последовательность, кортеж `tuple` — последовательность с неизменным составом. Индекс начинается с нуля, `-1` означает последний элемент. Индекс за границей вызывает `IndexError`.

Срез `[start:stop:step]` не включает `stop`; пропущенные границы означают начало или конец, а шаг по умолчанию равен 1. Шаг −1 идёт с конца к началу; срез `[::-1]` разворачивает последовательность. Срез за границей безопасен. Строки и кортежи нельзя менять по индексу; для нового текста создайте новую строку.

```python
name = "планер"
spans = [12, 22, 14, 20]
point = (12, 8)
print(name[0], spans[-1], point[1])  # п 20 8
print(spans[1:3])                   # [22, 14]
print(spans[::-1])                  # [20, 14, 22, 12]
print(spans[10:20])                 # []
```

`strip` убирает пробельные символы с краёв, `split` без аргументов разбивает по пробельным промежуткам, `split(",")` разбивает по заданному разделителю — запятой. `join` соединяет строки указанным разделителем. Сочетание `\n` внутри строки обозначает перевод строки. Вызов `объект.имя(...)` — метод: действие, связанное с объектом. Например, `spans.append(24)` меняет список и возвращает `None`.

```python
words = "  короткий дальний  ".strip().split()
print(words)               # ['короткий', 'дальний']
print(", ".join(words))    # короткий, дальний
```

## Множество и словарь

Множество `set` хранит уникальные элементы без позиционных индексов. Не полагайтесь на порядок его вывода. Словарь `dict` связывает уникальный ключ со значением; в Python он сохраняет порядок добавления ключей.

`in` проверяет принадлежность. В словаре он проверяет ключи. `get(key, default)` возвращает значение либо `default`, если ключ отсутствует; обращение `counts[key]` без существующего ключа вызывает `KeyError`. Пустое множество записывается `set()`, а `{}` — пустой словарь. Ключами могут быть строки, числа, кортежи неизменяемых элементов; списки ключами быть не могут.

```python
labels = ["дальний", "короткий", "дальний"]
unique = set(labels)
print(len(unique), "дальний" in unique)  # 2 True
counts = {}
for label in labels:
    counts[label] = counts.get(label, 0) + 1
print(counts)  # {'дальний': 2, 'короткий': 1}
```

## Действия с множествами

Метод `add` добавляет элемент, а повторное добавление ничего не меняет. `A & B` — пересечение (общие элементы), `A | B` — объединение (все элементы без повторов). `A.issubset(B)` проверяет, что каждый элемент `A` входит в `B`. Эти операции пригодятся при проверке разбиения данных.

```python
A = {1, 2}
A.add(2)
A.add(3)
B = {3, 4}
assert A & B == {3}
assert A | B == {1, 2, 3, 4}
assert {1, 2}.issubset(A)
```

## Включения и совместный обход

Включение строит новую коллекцию по правилу. Например, `[value ** 2 for value in [2, 3]]` даёт `[4, 9]`. Во включении словаря перед двоеточием записывается ключ, после — значение. Сначала прочитайте его как обычный цикл: «для каждого значения, удовлетворяющего условию, добавить выражение».

`enumerate` выдаёт пары «индекс, элемент»; два имени слева распаковывают пару. `zip(left, right)` выдаёт пары соответствующих элементов двух последовательностей, но заканчивается на более короткой. Если длины должны совпадать, проверьте их отдельно. Не превращайте включение в длинную цепочку условий: несколько строк цикла легче отладить.

```python
large = [value for value in spans if value >= 20]
squares = {value: value ** 2 for value in [2, 3]}
print(large, squares)  # [22, 20] {2: 4, 3: 9}
for index, label in enumerate(labels):
    print(index, label)
```

То же правило для списка можно записать обычным циклом:

```python
large_by_loop = []
for value in spans:
    if value >= 20:
        large_by_loop.append(value)
assert large_by_loop == large
```

## Ссылки и копии

Присваивание не копирует список: два имени могут указывать на один объект. `==` сравнивает значения, `is` — тождественность объектов; для проверки отсутствия значения используйте `is None`.

```python
original = [12, 22]
alias = original
alias[0] = 99
print(original)            # [99, 22]
copy = original.copy()
copy[0] = 12
print(original, copy)      # [99, 22] [12, 22]
```

Метод `copy` создаёт поверхностную копию: вложенные объекты остаются общими. Такой метод есть и у словаря. Кортеж тоже может содержать изменяемый список: нельзя заменить элемент кортежа, но можно изменить сам вложенный список.

## Разобранный пример: независимые строки таблицы

Вложенный список — список, чьи элементы сами являются списками. `table[0][0]` означает первый элемент первой строки. Умножение списка на 3 повторяет его элементы трижды; если элемент — список, повторяются ссылки. Во включении `_` — обычное имя для неиспользуемого значения перебора.

```python
bad = [[0, 0]] * 3
bad[0][0] = 1
print(bad)  # [[1, 0], [1, 0], [1, 0]]
good = [[0, 0] for _ in range(3)]
good[0][0] = 1
print(good) # [[1, 0], [0, 0], [0, 0]]
```

Умножение списка повторило ссылки на одну строку. Включение создало новую строку на каждом шаге. Для таблицы чисел независимую копию можно собрать как `[row.copy() for row in table]`: копируются и внешний список, и каждая строка.
