## Хеширование и подсчёт

Словарь и множество используют хеширование: по значению ключа вычисляют число — хеш, которое помогает выбрать место поиска. Если разные ключи получили одинаковый хеш, их дополнительно различают сравнением. Для обычных ключей проверка наличия занимает в среднем $O(1)$, но это не гарантия худшего случая. Частоты можно собрать за ожидаемое $O(n)$ время и $O(u)$ дополнительной памяти, где $u$ — число разных ключей.

## Разобранный пример: существует ли пара с заданной суммой

```python
def has_pair(values, target):
    seen = set()
    for value in values:
        if target - value in seen:
            return True
        seen.add(value)
    return False

assert has_pair([12, 22, 14], 34)
assert not has_pair([17], 34)
assert has_pair([17, 17], 34)
```

Для суммы 34 на входе `[12, 22, 14]` сначала ищем `34 - 12 = 22` в пустом множестве и добавляем 12. На следующем шаге ищем `34 - 22 = 12`; оно уже есть, поэтому возвращаем `True`.

Перед шагом `seen` содержит только предыдущие значения. Поэтому один объект не используется дважды; второе значение 17 может образовать пару с первым. Ожидаемое время $O(n)$, дополнительная память $O(n)$. Перебор пар использует $O(n^2)$ времени, но может обходиться постоянной дополнительной памятью, если нужен лишь ответ «да/нет».

## Разобранный пример: один проход вместо повторного поиска

Если для каждого слова заново считать его вхождения полным обходом, повторяется одна и та же работа. Словарь позволяет накапливать частоты за один проход. Память зависит от числа разных слов u, а не только от длины текста.

```python
words = ["кот", "кит", "кот", "кот"]
frequencies = {}
for word in words:
    frequencies[word] = frequencies.get(word, 0) + 1
assert frequencies == {"кот": 3, "кит": 1}
assert len(frequencies) == 2
```

Ожидаемое время O(n) относится к обычным ключам и операциям хеш-таблицы. Оно не означает, что любая операция над строкой имеет постоянную стоимость: вычисление хеша длинной новой строки тоже требует работы. В наших оценках длины ключей считаем ограниченными.

## Порядок действий — часть алгоритма

В поиске пары множество `seen` должно хранить только предыдущие значения. Если сначала добавить текущий элемент, он сможет образовать пару с собой. Проверка `[17]` и `[17, 17]` различает два варианта и помогает найти ошибку, которая не видна на примере `[12, 22]`.
