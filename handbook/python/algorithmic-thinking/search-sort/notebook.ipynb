## Поиск и сортировка

Линейный поиск в худшем случае смотрит все $n$ элементов. Сортировка располагает значения от меньшего к большему, сохраняя повторы. `sorted(values)` создаёт новый список; метод `values.sort()` меняет существующий и возвращает `None`, поэтому не пишите `values = values.sort()`. Сортировка Python имеет худшую временную оценку $O(n\log n)$.

```python
def find_index(values, target):
    for index, value in enumerate(values):
        if value == target:
            return index
    return None

assert find_index([4, 8, 4], 8) == 1
assert find_index([], 8) is None
ordered = sorted([8, 4, 4])
print(ordered)  # [4, 4, 8]
```

В отсортированном списке можно искать двоичным поиском: сравнить с серединой и оставить половину диапазона. После $k$ шагов остаётся примерно $n/2^k$ элементов, поэтому нужно $O(\log n)$ сравнений. Но предварительная сортировка стоит времени и меняет порядок. Для одного поиска она может быть лишней.

Проследим поиск 8 в `[1, 3, 5, 8, 10, 12, 14]`. Средний элемент равен 8 — ответ найден сразу. При поиске 12 сравниваем с 8, отбрасываем левую часть вместе с 8 и продолжаем в `[10, 12, 14]`. Если диапазон опустел, значения нет. Отбрасывать половину можно именно благодаря сортировке: слева не может скрываться число больше середины.

Модуль — набор готовых инструментов Python; запись `from bisect import bisect_left` подключит нужную функцию. `bisect_left` из стандартного модуля `bisect` ищет позицию вставки значения слева от равных ему элементов, сохраняя сортировку. Он не проверяет наличие значения: результат может быть равен длине списка, то есть указывать сразу за последним элементом. Поэтому сначала проверяем границу, затем равенство.

```python
from bisect import bisect_left

index = bisect_left(ordered, 8)
assert index < len(ordered) and ordered[index] == 8
```

## Разобранный пример: поиск отсутствующего значения

Для отсортированного `[4, 4, 8]` позиция вставки 7 равна 2, но элемент с индексом 2 равен 8. Поэтому проверка одной позиции недостаточна. Для значения 10 позиция равна длине списка: обращение по такому индексу вызвало бы ошибку.

```python
def contains_sorted(values, target):
    index = bisect_left(values, target)
    return index < len(values) and values[index] == target

assert contains_sorted(ordered, 4)
assert not contains_sorted(ordered, 7)
assert not contains_sorted(ordered, 10)
assert not contains_sorted([], 1)
```

Левая проверка через `and` защищает правую от выхода за границу. Условие работает и для пустого списка. Для повторяющихся значений `bisect_left` возвращает позицию первого из них.

## Выбор метода под задачу

Один запрос к несортированному списку удобно выполнить линейно. Много запросов к одному неизменному набору могут оправдать предварительную сортировку. Если нужны исходные индексы, сохраните их вместе со значениями: после сортировки позиции изменятся. Изменение данных между запросами тоже имеет стоимость, которую нельзя игнорировать.
