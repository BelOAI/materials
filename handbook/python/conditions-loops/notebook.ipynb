## Логические выражения

Сравнения `==`, `!=`, `<`, `<=`, `>`, `>=` возвращают `True` или `False`. `and` требует выполнения обоих условий, `or` — хотя бы одного, `not` меняет истину на ложь и наоборот. Скобки помогают явно задать порядок.

```python
mass = 12
print(10 <= mass < 15)                     # True
print(mass >= 10 and mass < 15)            # True
print(not (mass < 10 or mass >= 15))       # True
```

`and` не вычисляет правую часть, если левая уже ложна. Поэтому для списка с именем `values` проверка `len(values) > 0 and values[0] > 0` безопасно и при пустом списке. Обратный порядок такой защиты не даёт.

## Выбор одной ветви

В конструкции `if` / `elif` / `else` выполняется только первая подходящая ветвь. `elif` означает «иначе, если». Независимые `if` могут выполнить несколько действий.

```python
score = 70
if score >= 90:
    level = "отлично"
elif score >= 60:
    level = "зачёт"
else:
    level = "нужна практика"
print(level)  # зачёт
```

Ставьте более узкое условие раньше широкого: при проверке `score >= 60` первой оценка 95 тоже попадёт в «зачёт».

## Перебор и накопление

`range(start, stop, step)` задаёт целые числа от `start` до `stop`, не включая `stop`. По умолчанию начало равно 0, шаг равен 1. Шаг 0 запрещён. Отрицательный шаг позволяет идти назад.

`list` здесь превращает перебор в список для вывода; `+=` прибавляет к текущему значению. `continue` пропускает остаток текущего шага.

```python
print(list(range(2, 8, 2)))  # [2, 4, 6]
print(list(range(3, 0, -1))) # [3, 2, 1]
values = [12, -1, 22, 0, 14]
total = 0
count = 0
for value in values:
    if value <= 0:
        continue
    total += value
    count += 1
print(total, count)  # 48 3
```

После каждого шага `total` равен сумме уже просмотренных положительных значений — это утверждение помогает проверить алгоритм.

## Повторение до условия и досрочный выход

`while` проверяет условие перед каждым шагом. Если оно сразу ложно, тело не выполнится. Нужно понимать, какое изменение приближает завершение. `-=` вычитает из текущего значения. `break` завершает ближайший цикл; в следующем примере он остановит поиск на первом подходящем значении. `None` до начала поиска означает, что ответ ещё не найден.

```python
remaining = 13
steps = 0
while remaining > 0:
    remaining -= 4
    steps += 1
print(steps, remaining)  # 4 -3

first_large = None
for value in values:
    if value >= 20:
        first_large = value
        break
print(first_large)  # 22
```

Если подходящего элемента нет, сохраняется `None`. Это лучше, чем придумывать число, которое может оказаться настоящим ответом. Во вложенных циклах внутренний перебор выполняется заново для каждого внешнего элемента.

## Разобранный пример: пары подходящих размахов

Нужны пары разных объектов с суммой размахов 34. Метод — действие, которое вызывается через точку после имени объекта. Метод `pairs.append(...)` добавит элемент в конец списка, а запись `(i, j)` объединит два индекса в пару. Для индексов `i < j` каждая пара рассматривается ровно один раз.

```python
spans = [12, 22, 14, 20]
pairs = []
for i in range(len(spans)):
    for j in range(i + 1, len(spans)):
        if spans[i] + spans[j] == 34:
            pairs.append((i, j))
print(pairs)  # [(0, 1), (2, 3)]
```

Подробнее пары значений разберём в коллекциях. Например, при `i = 0` внутренний цикл проверяет `j = 1, 2, 3`; при `i = 1` — только `j = 2, 3`. Перебор всех пар может быть дорогим — оценивать его стоимость научимся в главе об алгоритмах.

### Задача 3

### Задача 4

```python
def count_steps(n, step):
    if type(n) != int or type(step) != int:
        raise ValueError("n и step должны быть целыми числами")
    if n < 0 or step <= 0:
        raise ValueError("Нужны n >= 0 и step > 0")
    count = 0
    while n > 0:
        n -= step
        count += 1
    return count

assert count_steps(13, 4) == 4
assert count_steps(0, 4) == 0
```
