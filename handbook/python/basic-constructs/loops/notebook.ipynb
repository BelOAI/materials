## Список и перебор

Список хранит последовательность значений в квадратных скобках: `[12, 22, 14]`. Пустой список — `[]`. `len(values)` даёт число элементов, `values[0]` — первый элемент: индексы начинаются с нуля. В цикле `for value in values` имя `value` по очереди получает каждый элемент. Тело цикла выделяется отступом.

Начнём с одного действия для каждого элемента. Имя после `for` выбирает программист; оно не является командой Python.

```python
for span in [12, 22, 14]:
    print(span)
```

Вывод — три строки: 12, 22, 14. Сначала `span` равно 12, затем 22, затем 14. После последнего элемента перебор завершается; для пустого списка тело не запускается.

## Перебор и накопление

`range(start, stop, step)` задаёт целые числа от `start` до `stop`, не включая `stop`. У `range(stop)` начало равно 0, шаг равен 1; у `range(start, stop)` задано начало, шаг равен 1. Например, `range(3)` выдаёт 0, 1, 2, а `range(2, 5)` — 2, 3, 4. Шаг 0 запрещён. Отрицательный шаг позволяет идти назад.

`list` здесь превращает перебор в список для вывода; `+=` прибавляет к текущему значению. `continue` пропускает остаток текущего шага.

```python
print(list(range(2, 8, 2)))  # [2, 4, 6]
print(list(range(3, 0, -1))) # [3, 2, 1]
values = [12, -1, 22, 0, 14]
total = 0
count = 0
for value in values:
    if value <= 0:
        continue
    total += value
    count += 1
print(total, count)  # 48 3
```

После 12 накопители равны `total=12`, `count=1`; значение −1 их не меняет; после 22 получаем 34 и 2; ноль пропускаем; после 14 — 48 и 3. После каждого шага `total` равен сумме уже просмотренных положительных значений — это утверждение помогает проверить алгоритм. Запись `total += value` здесь означает `total = total + value`.

## Повторение до условия и досрочный выход

`while` проверяет условие перед каждым шагом. Если оно сразу ложно, тело не выполнится. Нужно понимать, какое изменение приближает завершение. `-=` вычитает из текущего значения. `break` завершает ближайший цикл; в следующем примере он остановит поиск на первом подходящем значении. `None` до начала поиска означает, что ответ ещё не найден.

```python
remaining = 13
steps = 0
while remaining > 0:
    remaining -= 4
    steps += 1
print(steps, remaining)  # 4 -3

first_large = None
for value in values:
    if value >= 20:
        first_large = value
        break
print(first_large)  # 22
```

Если подходящего элемента нет, сохраняется `None`. Это лучше, чем придумывать число, которое может оказаться настоящим ответом. Во вложенных циклах внутренний перебор выполняется заново для каждого внешнего элемента.

## Разобранный пример: среднее положительных измерений

Для среднего нужны сумма и число подходящих элементов. Оба накопителя начинают с нуля. После каждого шага они описывают уже просмотренную часть списка.

```python
measurements = [10, -1, 20, 0]
total = 0
count = 0
for value in measurements:
    if value > 0:
        total += value
        count += 1
if count == 0:
    print("Нет положительных измерений")
else:
    print(total / count)  # 15.0
```

При переборе пустого списка тело цикла не выполнится ни разу. Для количества и суммы это даёт корректные нули, но среднее без измерений не определено. Переменную для суммы создавайте до цикла: обнуление внутри тела стирает накопленное.
