## Значения и имена

Числа, строки и логические значения имеют разные типы. Строка записывается в кавычках; `True` и `False` означают истину и ложь. В Python десятичная часть отделяется точкой. Знак `=` связывает имя слева со значением справа; такое имя называют переменной. Комментарий начинается с `#` и не выполняется. Сравнение `<` проверяет «меньше» и возвращает логическое значение.

````python
span = 22             # целое число: размах в сантиметрах
mass = 10.5           # число с дробной частью: масса в граммах
flight = "дальний"    # строка
is_light = mass < 15  # логическое значение
print(span, flight, is_light)
````

Вывод: `22 дальний True`. Имя можно использовать справа в новом выражении: сначала вычисляется правая часть с прежним значением `span`, затем имя связывается с результатом.

````python
span = span + 2
print(span)  # 24
````

`"22"` — строка, а `22` — число. `int("22")` превращает строку в целое число. Неверное преобразование, например `int("крыло")`, вызывает ошибку. Проверяйте смысл входных данных, а не только возможность преобразовать их.

Встроенные типы называются `int` (целые числа), `float` (числа с дробной частью), `str` (строки) и `bool` (логические значения). Вызов `type(value)` показывает тип значения. Имена переменных пишут без пробелов, начиная с буквы или подчёркивания; регистр важен: `span` и `Span` — разные имена. Выбирайте понятные имена, например `total`, а не `a1`.

В записи `int("22")` имя перед скобками задаёт действие, а значение внутри — его вход. Результат можно сохранить в переменной или использовать в другом выражении. `str(22)` делает обратное преобразование в текст, `int(12.5)` отбрасывает дробную часть к нулю и даёт 12.

## Арифметика и запись результата

Знаки `+`, `-`, `*` обозначают сложение, вычитание и умножение.

`/` делит, `//` даёт целую часть с округлением вниз, `%` — остаток, `**` возводит в степень. Для отрицательных чисел округление вниз отличается от отбрасывания дробной части: `-7 // 2` равно −4, `-7 % 2` равно 1. На ноль делить нельзя. Умножение и деление выполняются раньше сложения; скобки делают порядок явным.

````python
print(7 + 2, 7 - 2, 7 * 2)  # 9 5 14
print(7 / 2, 7 // 2, 7 % 2) # 3.5 3 1
print(2 ** 3)               # 8
print((2 + 3) * 4)          # 20
````

F-строка начинается с `f` перед кавычкой. В ней выражения в фигурных скобках заменяются значениями, а `:.2f` задаёт два знака после точки при выводе, не меняя само число:

````python
value = 3 / 4
print(f"Доля верных: {value:.2f}")  # Доля верных: 0.75
````

Дробные числа хранятся приближённо; способы сравнения с допуском изучим в NumPy.

## Ввод и вывод

`print` выводит значения через пробел и завершает строку. Функция `input` показывает приглашение в скобках и ожидает строку с клавиатуры. `float` преобразует строку с числом в дробное число; десятичный разделитель — точка. Следующий фрагмент **для ручного запуска**, поэтому в ноутбуке он дан как текст:

```text
span_text = input("Размах в сантиметрах: ")
span = float(span_text)
print("Размах:", span)
```

При вводе `12.5` вывод будет `Размах: 12.5`. `input` всегда возвращает строку: сложение `"12" + "3"` даёт `"123"`, а не 15. Сначала преобразуйте данные в число. `int` читает целое, `float` — дробное с точкой. Запись `int("12.5")` неверна; `float("12.5")` допустима.

## Если программа не запускается

Читайте последнюю строку сообщения: в ней тип ошибки и объяснение. `NameError` часто означает опечатку в имени или невыполненную ячейку; `ValueError` при `int("крыло")` — неподходящий текст; `ZeroDivisionError` — деление на ноль. Найдите указанную строку, проверьте значения и повторите запуск. Код после необработанной ошибки не выполняется. Подробно разбирать ошибки и проверять функции будем в отдельной главе.

## Разобранный пример: стоимость тетрадей

Три тетради по 2.50 рубля стоят 7.50 рубля. Сначала запишем входные данные именами, затем вычислим результат. Имя `price` хранит цену одной тетради, `count` — количество. Знак `==` сравнивает два значения и даёт `True` или `False`: `7 == 7` истинно. Он отличается от присваивания `=`. В решениях `assert условие` проверяет истинность условия и останавливает программу, если оно ложно. Успешный `assert` ничего не печатает, поэтому для показа ответа нужен `print`.

````python
price = 2.50
count = 3
total = price * count
assert total == 7.5
print(f"Стоимость: {total:.2f}")  # Стоимость: 7.50
````

Если ученик введёт количество строкой `"3"`, его нужно преобразовать через `int` до умножения. Проверьте себя: при цене 1.25 и количестве 4 ответ должен быть 5.00. Отображение двух знаков после точки меняет запись ответа, а не точность хранящегося числа.
