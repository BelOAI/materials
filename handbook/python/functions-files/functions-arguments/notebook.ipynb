## Как определить функцию

Функция объединяет действия под именем. Запись `def имя(параметры):` начинает её определение; тело записывается с отступом. Определение само не выполняет тело: для запуска нужен вызов `имя(аргументы)`. Ключевое слово `return` передаёт результат вызывающему коду и завершает вызов.

## Аргументы и результат

Параметры — имена в определении функции; аргументы — значения при вызове. Аргументы можно передать по позиции или по имени. Значение по умолчанию используется, если аргумент пропущен.

`sum` суммирует элементы, тройные кавычки задают строку документации. `return` немедленно завершает функцию; без него результат равен `None`. `print` показывает текст, но не заменяет возвращаемое значение.

```python
def count_above(values, threshold=20):
    """Число элементов не меньше порога."""
    return sum([1 for value in values if value >= threshold])

print(count_above([12, 22, 24]))                 # 2
print(count_above([12, 22, 24], threshold=23))   # 1
```

При первом вызове параметр `threshold` равен 20. Включение создаёт `[1, 1]`: по единице за 22 и 24. `sum` складывает их и возвращает 2. При втором вызове порог 23, подходит только 24. Для пустого входа получается `sum([])`, то есть 0.



## Контракт функции

`raise ValueError("сообщение")` останавливает текущий вызов, сообщая о недопустимом значении. Обработку таких ошибок изучим в уроке проверки программ. Здесь важно проверить вход до деления, чтобы причина ошибки была понятной.

Контракт описывает допустимые входы, результат и изменения данных. Например, функция среднего получает непустой список чисел и возвращает число. Функция счёта подходящих элементов допускает пустой список и возвращает 0. Названия и примеры помогают отличить эти случаи до написания кода.

```python
def average(values):
    if len(values) == 0:
        raise ValueError("Нужны измерения")
    return sum(values) / len(values)

assert average([10, 14, 18]) == 14
```

Для `[10, 14, 18]` сумма равна 42, длина — 3, возвращаемое среднее — 14. Для пустого списка срабатывает `raise`, и до деления выполнение не доходит.

## Разобранный пример: отделяем вычисление от вывода

```python
def rectangle_area(width, height=1):
    return width * height

area = rectangle_area(3, height=4)
assert area == 12
print(area)
```

Параметр по умолчанию `height=1` позволяет вызвать `rectangle_area(3)`. Вызов с именем аргумента показывает, что означает число 4. Функция возвращает результат, а печатает его вызывающая программа: так число можно использовать в дальнейших вычислениях и тестах.

```python
def sum_large(values, threshold=20):
    total = 0
    for value in values:
        if value >= threshold:
            total += value
    return total
assert sum_large([12, 22, 24]) == 46
assert sum_large([]) == 0
assert sum_large([12, 22, 24], threshold=23) == 24
```
