## Локальные и внешние имена

Имя, созданное внутри функции, локально: снаружи к нему нельзя обратиться по этому имени. Внешнее имя можно читать, но присваивание внутри функции обычно создаёт локальное имя. Лучше передавать нужные данные аргументами, чтобы поведение не зависело от скрытого состояния ноутбука.

```python
threshold = 20
def shifted(value):
    threshold = 5
    return value + threshold
assert shifted(2) == 7
assert threshold == 20
```

Локальный `threshold` не заменил внешний. Если функция читает внешнюю переменную, изменение этой переменной перед вызовом может изменить результат той же функции. Такой зависимости стоит избегать в вычислительных задачах.

## Изменяемые аргументы

Функция получает ссылку на объект. Изменение переданного списка через `append` видно вызывающему коду. Для функции, которая должна сохранять вход, создайте новый список.

Значения по умолчанию вычисляются один раз при определении функции. Поэтому `def add(value, items=[])` опасно: вызовы будут делить один список. Используйте `None`:

```python
def add(value, items=None):
    if items is None:
        items = []
    items.append(value)
    return items

assert add(1) == [1]
assert add(2) == [2]
```

## Разобранный пример: сохраняем вход

```python
def add_copy(values, value):
    result = values.copy()
    result.append(value)
    return result

source = [1, 2]
answer = add_copy(source, 3)
assert source == [1, 2]
assert answer == [1, 2, 3]
```

Функция, меняющая вход, тоже может быть правильной, если это явно предусмотрено контрактом. При обработке данных важно согласовать ожидания: вызывающий код может использовать исходный список в следующем эксперименте.

## Новое присваивание параметра

```python
def replace_local(values):
    values = [99]
    return values

source = [1, 2]
assert replace_local(source) == [99]
assert source == [1, 2]
```

Параметр связывается с объектом при вызове. Новое присваивание меняет эту локальную связь, а `append` или присваивание по индексу изменяет сам переданный объект. Это та же модель ссылок, которую мы разобрали для списков.

## Почему пустой список по умолчанию накапливает ответы

В следующем примере список создаётся при исполнении строки `def`. При втором вызове Python не создаёт его заново.

```python
def collect_shared(value, items=[]):
    items.append(value)
    return items

first = collect_shared(1)
second = collect_shared(2)
assert first is second
assert first == [1, 2]
```

Даже сохранённый первый результат изменился, потому что это тот же список. Вариант с `None` выше создаёт новый список внутри каждого вызова без переданного `items`. Если же передать свой список явно, функция добавит элемент именно в него.
