## Пути и текст

Модуль — набор готовых функций и типов; `import` делает его доступным программе. `Path` из модуля `pathlib` представляет путь. Относительный путь отсчитывается от текущей рабочей папки, которая не всегда совпадает с папкой ноутбука. `Path.cwd()` показывает её: после `from pathlib import Path` выполните `print(Path.cwd())` и сравните показанную папку с местом, куда положили файл. Абсолютный путь начинается от корня файловой системы и не зависит от текущей папки. Оператор `/` для путей добавляет часть пути.

Текстовые файлы хранят байты; кодировка объясняет, как превратить их в символы. Для наших файлов явно задаём UTF-8. Режим `"r"` читает, `"w"` записывает и стирает прежнее содержимое. Конструкция `with ... as stream` открывает ресурс и даёт ему имя `stream` внутри блока; по окончании блока файл закрывается даже при исключении. `path.open(...)` открывает файл, `stream.write(text)` записывает текст, `stream.read()` читает весь текст. Последовательность `\n` внутри строки обозначает перевод строки.

`from module import name` подключает только указанное имя из модуля. `TemporaryDirectory` из `tempfile` создаёт временную папку; `with TemporaryDirectory() as folder` даёт её путь и удаляет папку при выходе из блока.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as folder:
    path = Path(folder) / "note.txt"
    with path.open("w", encoding="utf-8") as stream:
        stream.write("Размах: 22 см\n")
    with path.open(encoding="utf-8") as stream:
        text = stream.read()
    assert text.strip() == "Размах: 22 см"
print(text.strip())  # Размах: 22 см
```

Пример не перезаписывает ученические файлы. Для большого файла перебирайте строки `for line in stream`, а не загружайте весь текст через `read`.

## Разобранный пример: числа по строкам

Файл можно перебирать построчно. Каждая строка обычно заканчивается `\n`; `strip` удаляет крайние пробельные символы. Пустую строку нельзя превратить в число, поэтому для нашего формата явно договоримся пропускать пустые строки.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "numbers.txt"
    with path.open("w", encoding="utf-8") as stream:
        stream.write("12\n\n18\n")
    values = []
    with path.open(encoding="utf-8") as stream:
        for line in stream:
            cleaned = line.strip()
            if cleaned != "":
                values.append(int(cleaned))
if len(values) == 0:
    raise ValueError("В файле нет числовых измерений")
assert values == [12, 18]
average = sum(values) / len(values)
assert average == 15
```

Это правило относится к конкретному формату. Для таблицы с несколькими полями пустое поле может быть важным пропуском; автоматически удалять его нельзя. Чтение всего файла через `read` удобно для небольшого текста, построчный обход экономит память на больших файлах.

## Причины ошибок чтения

Отсутствующий файл, неправильная кодировка и неверный формат числа — разные проблемы. Сначала проверьте рабочую папку через `Path.cwd()`, затем имя файла, затем договорённость о содержимом. Режим `"w"` заменяет прежнее содержимое, поэтому учебные записи делаем во временной папке.

Пустой файл или файл только из пустых строк оставит `values` пустым. Поэтому до деления проверяем его длину: отсутствие измерений не заменяем средним 0. Проверка известного списка `[12, 18]` подтверждает пример; условие перед делением задаёт осмысленную обработку пустого результата.
