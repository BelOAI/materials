## Строка как последовательность

Строка хранит текст в кавычках. Индекс начинается с нуля, отрицательный индекс отсчитывается с конца: `text[-1]` — последний символ. `len(text)` даёт число символов. Обращение за границу вызывает `IndexError`; у пустой строки нет первого символа.

Срез `[start:stop:step]` выбирает символы от `start` до `stop`, не включая правую границу. Пропущенные границы означают начало или конец, шаг по умолчанию равен 1. Отрицательный шаг идёт назад.

```python
text = "планер"
assert text[0] == "п"
assert text[-1] == "р"
assert text[1:4] == "лан"
assert text[::-1] == "реналп"
assert text[20:30] == ""
```

Строка неизменяема: нельзя присвоить `text[0] = "П"`. Чтобы получить новый текст, соедините части или используйте метод. Метод — действие, вызываемое через точку после объекта.

## Очистка и разбиение

`strip` убирает пробельные символы с краёв, `lower` переводит буквы в нижний регистр, `split` без аргументов разбивает по пробельным промежуткам. `split(",")` использует конкретный разделитель. Сочетание `\n` в строке обозначает перевод строки.

```python
raw = "  Кот   КИТ\n"
words = raw.strip().lower().split()
assert words == ["кот", "кит"]
assert "a,,b".split(",") == ["a", "", "b"]
```

CSV — текстовый формат таблиц: поля обычно разделены запятыми, а поле с запятой внутри заключают в кавычки. Например, `a,"крыло, большое",22` — три поля, хотя запятых три. Подробно читать такие файлы научимся позже.

Явный разделитель сохраняет пустые поля. Это важно для данных: пустое значение между запятыми не должно бесследно исчезнуть. CSV с кавычками будем читать специальным модулем.

## Разобранный пример: единая запись слов

Метод `join` соединяет список строк указанным разделителем. Нельзя соединить числа без предварительного преобразования `str`. Операция `in` проверяет, встречается ли подстрока.

```python
raw = "  Короткий   ДАЛЬНИЙ "
normalized = " ".join(raw.lower().split())
assert normalized == "короткий дальний"
assert "дальний" in normalized
```

Цепочка читается слева направо: `raw.lower()` даёт `"  короткий   дальний "`, вызов `.split()` — список `["короткий", "дальний"]`, а `" ".join(...)` вставляет между словами один пробел. Исходная строка не изменилась. Удаление пробелов внутри имени или метки требует отдельного решения: иногда оно меняет смысл данных.
