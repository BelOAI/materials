## Ссылки и копии

Присваивание не копирует список: два имени могут указывать на один объект. `==` сравнивает значения, `is` — тождественность объектов; для проверки отсутствия значения используйте `is None`.

```python
original = [12, 22]
alias = original
alias[0] = 99
print(original)            # [99, 22]
copy = original.copy()
copy[0] = 12
print(original, copy)      # [99, 22] [12, 22]
```

Метод `copy` создаёт поверхностную копию: вложенные объекты остаются общими. Такой метод есть и у словаря. Кортеж тоже может содержать изменяемый список: нельзя заменить элемент кортежа, но можно изменить сам вложенный список.

## Разобранный пример: независимые строки таблицы

Вложенный список — список, чьи элементы сами являются списками. `table[0][0]` означает первый элемент первой строки. Умножение списка на 3 повторяет его элементы трижды; если элемент — список, повторяются ссылки. Во включении `_` — обычное имя для неиспользуемого значения перебора.

```python
bad = [[0, 0]] * 3
bad[0][0] = 1
print(bad)  # [[1, 0], [1, 0], [1, 0]]
good = [[0, 0] for _ in range(3)]
good[0][0] = 1
print(good) # [[1, 0], [0, 0], [0, 0]]
```

Умножение списка повторило ссылки на одну строку. Включение создало новую строку на каждом шаге. Для таблицы чисел независимую копию можно собрать как `[row.copy() for row in table]`: копируются и внешний список, и каждая строка.

## Изменение объекта и новое присваивание

Присваивание имени нового списка не меняет объект, который раньше был с ним связан. Изменение элемента через индекс меняет сам объект и видно через все его имена.

```python
a = [1, 2]
b = a
b = [9, 9]
assert a == [1, 2]
b = a
b[0] = 9
assert a == [9, 2]
```

Для сравнения значений используйте `==`. Оператор `is` отвечает на другой вопрос: один ли это объект. Для проверки отсутствия значения обычно пишут `is None`. Не используйте `is` вместо равенства чисел или строк.

## Разобранный пример: сохраняем исходную таблицу

```python
source = [[12, 8], [22, 10]]
working = [row.copy() for row in source]
working[0][0] = 15
assert source == [[12, 8], [22, 10]]
assert working == [[15, 8], [22, 10]]
```

Такой способ достаточен для таблицы с двумя уровнями списков и числами. Если внутри строки снова лежит список, понадобится отдельно решить, какие уровни копировать. Перед обработкой данных сформулируйте, разрешено ли менять вход; это часть поведения программы.

## Где заканчивается поверхностная копия

Проверим именно вложенный объект, а не только внешний список.

```python
source = [[1], [2]]
shallow = source.copy()
assert shallow is not source
assert shallow[0] is source[0]
shallow[0].append(9)
assert source == [[1, 9], [2]]
```

Внешних списков два, но первая строка у них общая. Поэтому слово «копия» без указания уровня ещё не обещает независимость данных.

### Задача 6

Да: копия внешнего списка и копия строки новые, но самый внутренний список `[1]` остаётся общим. Поэтому `a` тоже станет `[[[1, 2]]]`. Для этой структуры нужно копировать каждый изменяемый уровень:

```python
a = [[[1]]]
b = []
for row in a:
    new_row = []
    for cell in row:
        new_row.append(cell.copy())
    b.append(new_row)
b[0][0].append(2)
assert a == [[[1]]]
assert b == [[[1, 2]]]
```

Числа внутри последних списков неизменяемы, поэтому дальше копировать нечего. Способ копирования зависит от структуры данных, а не только от вызова `.copy()`.
