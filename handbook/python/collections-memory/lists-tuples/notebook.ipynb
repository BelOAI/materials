## Упорядоченные данные

Список `list` хранит элементы в заданном порядке. Значения могут повторяться. Как у строки, доступны индексы, срезы и `len`, но список можно менять. Метод `append` добавляет элемент в конец и возвращает `None`; не присваивайте этот результат вместо списка.

```python
values = [12, 22, 14]
values[0] = 13
values.append(24)
assert values == [13, 22, 14, 24]
assert values[1:3] == [22, 14]
```

Кортеж `tuple` имеет неизменный состав: `(12, 8)` удобно хранит одну пару координат. Пустой кортеж — `()`, кортеж одного элемента — `(3,)`. Запятая отличает его от числа в скобках. Два имени слева распаковывают пару.

```python
point = (12, 8)
span, mass = point
assert span == 12 and mass == 8
```

## Включения и обход с индексом

Включение строит новый список по правилу: выражение, затем `for`, затем необязательное условие. Сначала удобно написать обычный цикл и проверить его смысл.

```python
values = [-2, 0, 3, 4]
result = []
for value in values:
    if value > 0:
        result.append(value ** 2)
short_result = [value ** 2 for value in values if value > 0]
assert result == short_result == [9, 16]
```

`enumerate` выдаёт пары «индекс, элемент», начиная с индекса 0. Например, `enumerate([12, 22])` даёт сначала `(0, 12)`, затем `(1, 22)`. В строке `for index, value in ...` два имени распаковывают эту пару, как `span, mass = point` выше. `zip` объединяет соответствующие элементы двух последовательностей, но заканчивается на более короткой. Перед сравнением ответов длины нужно проверять явно.

## Разобранный пример: доля верных ответов

```python
actual = [0, 1, 0, 1]
predicted = [0, 1, 1, 1]
assert len(actual) == len(predicted) and len(actual) > 0
correct = 0
for expected, answer in zip(actual, predicted):
    if expected == answer:
        correct += 1
assert correct / len(actual) == 0.75
```

Здесь нельзя считать качество на пустом списке: знаменатель будет нулём. Проверка равных длин защищает от потерянных ответов. Для поиска ошибок полезно одновременно выводить номер объекта и значения; номер можно получить через `enumerate`.

### Задача 6

При распаковке сначала вычисляется правая пара `(2, 1)`, затем её элементы записываются в `a` и `b`. Поэтому значения меняются местами. Последовательные команды сначала делают `a` равным 2, затем копируют это уже новое значение в `b`: получаются 2 и 2. Исходное значение `a` потеряно. Для любых разных исходных чисел обе переменные станут равны прежнему `b`.

```python
a, b = 1, 2
a, b = b, a
assert (a, b) == (2, 1)
a, b = 1, 2
saved = a
a = b
b = saved
assert (a, b) == (2, 1)
```
