## Когда важна принадлежность

Множество `set` хранит уникальные элементы без позиционных индексов. Пустое множество создают `set()`, запись `{}` означает словарь. Преобразование списка через `set(values)` удаляет повторы; порядок при этом не обещается. Для проверки наличия используйте `in`, для количества — `len`.

```python
labels = ["дальний", "короткий", "дальний"]
unique = set(labels)
assert len(unique) == 2
assert "дальний" in unique
```

Метод `add` добавляет элемент; повторное добавление ничего не меняет. Числа и строки можно хранить в множестве, списки — нельзя: элемент должен подходить для хеширования, способа поиска по значению. Подробнее оценим этот поиск в алгоритмах.

## Пересечение и объединение

`A & B` содержит общие элементы, `A | B` — все элементы без повторов, `A - B` — элементы A, которых нет в B. `A.issubset(B)` проверяет, что каждый элемент A входит в B.

```python
A = {1, 2, 3}
B = {3, 4}
assert A & B == {3}
assert A | B == {1, 2, 3, 4}
assert A - B == {1, 2}
assert {1, 2}.issubset(A)
```

Вычитание зависит от порядка: `B - A` равно `{4}`. На пустых множествах операции тоже определены: пересечение пусто, объединение с пустым множеством сохраняет исходное.

## Разобранный пример: проверка разбиения объектов

Пусть числа — идентификаторы четырёх объектов, то есть уникальные номера. Разделим их на учебную часть `train` и проверочную `test`. Позже программа-модель будет учиться на примерах из первой части и проверяться на второй. Одни и те же объекты не должны попасть в обе части. Пересечение частей должно быть пустым, а объединение — содержать все ожидаемые объекты.

```python
train = {1, 2, 3}
test = {4}
all_ids = {1, 2, 3, 4}
assert train & test == set()
assert train | test == all_ids
```

Такая проверка работает по идентификаторам, а не по позициям строк. Она не доказывает, что под разными идентификаторами не скрываются одинаковые реальные объекты; смысл ключей тоже нужно понимать.
